# The dict story so far

1. Dicts are key-value pairs
    - Every key has a value, every value has a key
    - When you run `len` on a dict, you get the number of pairs, not the number of keys + values.
2. You create a dict with `{}`
    - Empty dict is `{}`
    - You can create one with `{'a':10, 'b':20, 'c':30}`
3. Dicts are mutable!
    - You can replace an existing value with assignment (e.g., `d['a'] = 999`)
    - You can add a new key-value pair with assignment, assuming the key wasn't there already (e.g., `d['newthing'] = 'hello!'`)
    - You can remove a key-value pair by invoking `dict.pop`, naming the key -- you get the value back (e.g., `d.pop('a')`)

# Dict paradigms

1. Pretend it's immutable: Define a dict at the start of a program, and never change it -- not the keys, and not the values. Use it as a read-only dictionary in your program.
2. Modify the values, but not the keys: Define a dict at the start of a program with keys and default/basic values. Over the course of the program, update the values as needed -- values can be integers, lists, or anything else, including another dict.
