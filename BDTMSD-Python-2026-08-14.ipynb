# The dict story so far

1. Dicts are key-value pairs
    - Every key has a value, every value has a key
    - When you run `len` on a dict, you get the number of pairs, not the number of keys + values.
2. You create a dict with `{}`
    - Empty dict is `{}`
    - You can create one with `{'a':10, 'b':20, 'c':30}`
3. Dicts are mutable!
    - You can replace an existing value with assignment (e.g., `d['a'] = 999`)
    - You can add a new key-value pair with assignment, assuming the key wasn't there already (e.g., `d['newthing'] = 'hello!'`)
    - You can remove a key-value pair by invoking `dict.pop`, naming the key -- you get the value back (e.g., `d.pop('a')`)

# Dict paradigms

1. Pretend it's immutable: Define a dict at the start of a program, and never change it -- not the keys, and not the values. Use it as a read-only dictionary in your program.
2. Modify the values, but not the keys: Define a dict at the start of a program with keys and default/basic values. Over the course of the program, update the values as needed -- values can be integers, lists, or anything else, including another dict.

# Paradigm 3 -- start with nothing

If you don't know what keys or values you'll want to store in the dict, but you do know what you want to do with them, then you can do the following:

- Start with a completely empty dict
- When you get a new key-value pair, check if the key is already in the dict
    - If so, then update the value (usually adding to it)
    - If not, then add a new key-value pair, often with the key and 0, an empty list, 1, or something like that

This is perfect if you know the category if keys and values you'll get, but not the actual keys or values themselves.



# Example: Count characters

I want to use a dict to count how often each character appears in a user's input.

1. We'll ask the user to enter a string.
2. We'll go through that string, one character at a time.
3. Each character will be a key in our dict. The values will be ints, counting the number of times that character appears.

```python
counts = {}

text = input('Enter some text: ').strip()

for one_character in text:        # go through each character
    counts[one_character] += 1    # add 1 to the value associated with the key, our character

print(counts)
```

There are a few ways that we could fix this:

1. Define the dict with all characters as keys, and then 0s as the values. However, there are so many that this would result in massive memory bloat, and for what?
2. Use a special-purpose dictionary (and they do exist!) that lets you do `+=`, adding if needed.
3. Start empty, and then either update the value or add the key-value pair. We're going to that here.

```python
counts = {}

text = input('Enter some text: ').strip()

for one_character in text:            # go through each character
    if one_character in counts:       # we're checking if the character is a key in the "counts" dict -- have we seen it before?
        counts[one_character] += 1    # add 1 to the value associated with the key, our character
    else:
        counts[one_character] = 1     # add a new key-value pair to our "counts" dict, starting off with 1 occurrence

print(counts)
```

# Exercise: Rainfall

1. Define an empty dict, `rainfall`.  We will, over time, fill this dict up with new keys (city names, strings) and new values (integers, mm of rain that fell).
2. Ask the user, repeatedly, to enter the name of a city.
    - If you get an empty string, then the user wants to stop -- exit the loop and stop asking
3. Assuming you got a city name, ask the user for how much rain fell.
    - We can assume that the user will enter an integer.
4. If we've seen this city before, then add the new rainfall amount to the existing one.
5. If we haven't seen this city before, then add a new key-value pair, with this rainfall report.
6. Print the dict with all of the cities and rainfall amounts.

https://practice.lernerpython.com/classroom/ce06963a8d/ex-23

```python
rainfall = {}

while True:
    city_name = input('City: ').strip()

    if city_name == '':   # empty string?
        break

    mm_rain = input('Rain: ').strip()
    mm_rain = int(mm_rain)  # we assume our users only give us the best inputs!

    if city_name in rainfall:            # if we have seen this city before...
        rainfall[city_name] += mm_rain   # ... add to existing rainfall
    else:
        rainfall[city_name] = mm_rain    # ... adds a new key-value pair, with the city name and mm_rain

print(rainfall)        
```

# Iteration and dicts

We've seen that we can use a `for` loop to iterate over many different types of data:

- String: We get each character
- List or tuple: We get each element
- Dict: What do we get?

```python
d = {'a':10, 'b':20, 'c':30}

for one_thing in d:  # iterating over a dict gives you the *keys*
    print(one_thing)
```

```python
for one_key in d:
    print(f'{one_key}: {d[one_key]}')   
```
