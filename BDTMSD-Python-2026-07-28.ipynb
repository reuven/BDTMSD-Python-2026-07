# This can clash with the real world

If you need a string, and the string needs to include `\` characters, then you have to be careful!

People who use Windows use `\` to separate folder names.

```python
path = 'c:\abcd\ef\ghijk\lmn.doc'
```

```python
print(path)
```

Another special character in Python is `\a`, which is known as the "alarm bell." When our string contains `\a`, Python converts that into the alarm-bell character, and then when you print the string, it sounds.

If there is a special character that starts with `\`, then Python will convert it into that special character in your string.

As of recently, if your string contains `\` and then a character that isn't special, Python will give you a *warning*. 

```python
# how can we solve this?
# answer is easy: we double the \ -- that "escapes" the backslash, and thus makes it normal
# \a is the alarm bell
# but \\a is a \ character, followed by a

s = 'a\\b'
print(s)
```

```python
len(s)
```

```python
# one solution: double every backslash that you want to be taken literally.
path = 'c:\\abcd\\ef\\ghijk\\lmn.doc'

print(path)
```

```python
# if we just look at the definition of the string in Jupyter, without print, we'll see the doubles

path
```

```python
# that's kind of annoying! 
# Python comes to the rescue with "raw strings"
# - you put an r before the opening quote
# - every \ in the string is automatically doubled

# Basically, a raw string is a "what you see is what you get" string, without special characters

path = r'c:\abcd\ef\ghijk\lmn.doc'   # exactly the same as I did in cell #9
path
```

- What if I don't know the length of the string?
- This violates the DRY rule -- "don't repeat yourself!"

The basic idea in programming is: If you write something more than once, you're probably doing it wrong.

If you find yourself repeating things, you should find another way. Programmers have many ways to avoid repetition.

One big technique is *loops*, where we ask the computer to repeat things on our behalf. Python has *two* types of loops:

- `for`
- `while`

That's it! We probably use `for` loops in 3/4 of the cases.

# `for` loops

A `for` loop mean: I want to *iterate* over every character in a string. I then want to do something (or have the opportunity to do something) with each of those characters.

```python
s = 'abcd'

for one_character in s:
    print(one_character)
```

# Syntax of a `for` loop

1. We start with `for`
2. Next comes the *loop variable*. This is the variable to which a new value will be assigned with each iteration. The name of the loop variable has *NO* effect whatsoever on what we get with each iteration.
3. After that comes the keyword `in`, and here it's just separating the loop variable from the iterable.
4. Next comes the *iterable* value, the thing over which we're running our loop -- currently, a string. When I say that I want to "iterate over ____," that means that `___` will be at the end of the first line of the `for` loop.
5. At the end of that line comes `:`
6. Then we have the loop body, indented. The loop body can contain ANY PYTHON CODE AT ALL. It will be executed once for each iteration that we do through the loop. The loop body can contain as much or as little code as you want.
7. Once the loop is finished executing, the program continues after the loop body.

# `for` loops as conversations

1. `for` turns to the value at the end of the line (`s`, in this case) and asks: Are you iterable? Meaning, do you know how to act in a `for` loop?
    - If the answer is "no," then the loop exits with an error.
2. `for` turns again to the iterable value, and asks for its next value.
    - If there are no more values, the loop exits.
3. The value we got is assigned to the loop variable.
4. The loop body now runs.
5. When we get to the end of the loop body, we go back to step 2.

# This seems weird!

- Where is the index?
- How does the loop know what we'll be getting?
- How does the loop know how many iterations we'll have?

Answer: The loop knows nothing -- it just knows how to ask for the next value and assign it to the loop variable.
