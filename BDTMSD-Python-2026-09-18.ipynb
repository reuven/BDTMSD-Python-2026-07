# `NaN`

`NaN` is the Pandas way of printing something that is also/often known as `nan`. Both of these are the same, and both stand for "not a number."

The idea is that there are many times when you don't have data, or you do have data that has been corrupted. You don't want to replace those values with 0 or -9999 or something like that, since throw off your calculations.

Instead, you want to have something that cannot be mistaken for a number! 

```python
import numpy as np
import pandas as pd
from pandas import Series, DataFrame
```

```python
# Pandas uses NumPy under the hood -- and all of its data and all dtypes are actually NumPy dtypes
```

```python
np.nan
```

```python
type(np.nan)  # what kind of value is this?
```

```python
np.nan + 5
```

```python
10 * np.nan
```

```python
np.NaN  # Pandas style of writing NaN... this worked until about a year or so ago
```

The notion that `np.nan` can be used in any calculation, but immediately makes that calculation into `nan`, comes from the mathematicians who drove the NumPy implementation, wanting it to be "pure" and mathematically accurate.

However, Pandas is driven not by mathematical purity, but rather by pragmatism. In the real world, we normally just remove all of the NaN values in order to calculate what we want.

Pandas thus throws out nan values, rather than letting them dictate what will happen.

# Nullable types

What we're seeing here is some of the weirdness/frustration of dealing with NaN:

- If I have an integer column
- Some of the data is missing
- Why must I switch it to float? Floats are slower and don't have as much of a range

This is necessary because NumPy's dtypes are not *nullable*. `np.nan` is a float, and cannot be anything else. If you have an `np.nan` value in your series, the entire series must be float *or* becomes `object`, which means it basically can contain anything, but without any specific abilities as numbers, strings, or dates.

NumPy is not about to change. But people in the Pandas world have been very frustrated by this for a while.

There are two basic solutions (neither perfect):

1. Use Pandas "extension types," which are basically wrappers around the NumPy dtypes. You can tell the difference because the dtype names all start with Capital Letters. So you have `Int64` rather than `int64`.
2. Don't store things in the back end with NumPy. Rather, use PyArrow, whose values are all nullable. PyArrow will be introduced as a stable back end and probably replace NumPy at some point in the near future.. but for now, it's still experimental.

In both of these cases, you will then not use `np.nan` as your "unknown" value, but rather `pd.NA`, a Pandas-specific "not available" value.

```python
s = Series([10, 20, 30, pd.NA, 50, 60, pd.NA, 80, 90], dtype='Int64')  # capitalized, so an "extension" type
s
```

So far as I can tell, extension types never took off. The idea wasn't bad, but they don't give you any speed advantage.

Arrow/PyArrow is increasingly popular in the data world in general, in part because of its lightning speed.
