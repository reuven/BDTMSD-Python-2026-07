# `NaN`

`NaN` is the Pandas way of printing something that is also/often known as `nan`. Both of these are the same, and both stand for "not a number."

The idea is that there are many times when you don't have data, or you do have data that has been corrupted. You don't want to replace those values with 0 or -9999 or something like that, since throw off your calculations.

Instead, you want to have something that cannot be mistaken for a number! 

```python
import numpy as np
import pandas as pd
from pandas import Series, DataFrame
```

```python
# Pandas uses NumPy under the hood -- and all of its data and all dtypes are actually NumPy dtypes
```

```python
np.nan
```

```python
type(np.nan)  # what kind of value is this?
```

```python
np.nan + 5
```

```python
10 * np.nan
```

```python
np.NaN  # Pandas style of writing NaN... this worked until about a year or so ago
```

The notion that `np.nan` can be used in any calculation, but immediately makes that calculation into `nan`, comes from the mathematicians who drove the NumPy implementation, wanting it to be "pure" and mathematically accurate.

However, Pandas is driven not by mathematical purity, but rather by pragmatism. In the real world, we normally just remove all of the NaN values in order to calculate what we want.

Pandas thus throws out nan values, rather than letting them dictate what will happen.

```python
# some fun things with np.nan

type(np.nan)
```
