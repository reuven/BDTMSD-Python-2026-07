# What is going on with these data structures?

In theory, you could write every program you would want using just 1s and 0s. We build these higher-level data structures so that we can think at a more human level about what we're doing.

So far, we've talked about a bunch of relatively simple data structures. 

- Integers -- whole numbers. We use these for counting, but we also use them for indexing.
- Floats -- numbers with a decimal point. We use these for accurate calculations.
