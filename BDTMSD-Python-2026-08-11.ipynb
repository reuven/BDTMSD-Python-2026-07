# Dictionaries

Python calls them dicts, but they exist in other languages, too:

- Hashes
- Hash maps
- Maps
- Hash tables
- Key-value pairs
- Name-value pairs
- Associative arrays

The basic idea of a dict is that you don't store one thing, as with a list. Rather, you store *two* things, the key and the value. The key comes in place of the index in a list.

In a list, you can store whatever you want, but the index is determined by when you added the value to the list. The index starts with 0, then 1, then 2, until `len(the_list) - 1`.

In a dict, you can determine what the key (the index, basically) is! If you want to use people's ID numbers, you can do that. They don't need to be sequential. Moreover, they don't need to be integers.

# Basic rules and syntax for dicts

1. We create a dict with `{}`.
2. Each key-value pair has a `:` between the key and the value.
3. Pairs are separated from one another with `,`, as in a list or tuple.

## Some rules for dicts in general

1. Every key has a value, and every value has a key.
2. A key can be anything whatsoever, so long as it is immutable. (Basically, meaning that they have to be strings or numbers.)
3. The keys in a given dict must be unique. A key cannot repeat itself in the same dict, although it can be used in multiple dicts.
4. There are no restrictions whatsoever on the values. They can be ints, strings, lists, tuples, or even other dicts.
5. Everything in a dict goes through the keys. You can get the value from a key, but you cannot (easily) get a key from a value. One reason for this is that values can repeat, whereas keys cannot.

```python
d = {'a':10, 'b':20, 'c':30}    # this dict has 3 key-value pairs
```

```python
type(d)   # what kind of value is d referring to?
```

```python
len(d)   # how many pairs are there in this dict?
```

```python
# to retrieve a value, put the key inside of `[]`

d['a']  # retrieve the value associated with the key 'a'
```

```python
d['A']  # capital A?
```

```python
d['c']
```

If we try to retrieve a key that doesn't exist, then we get a `KeyError` exception. How can we avoid that? We can first ask the dict if a key is there.

We do that with the `in` operator. With strings, lists, and tuples, `in` returned `True` if the element was in the sequence. In the case of a dict, it *only* checks the keys.

```python
'a' in d   # is 'a' a key in dict d?
```

```python
'hello' in d   # is 'hello' a key in d?
```

```python
# I can put a string (or any other key) in a variable, and then retrieve with that

k = 'b'

if k in d:
    print(f'Yes, the value for {k} is {d[k]}')
else:
    print(f'No key {k} in our dict')
```

Can a key have more than one value?

- Each key can only have one value
- But that value can be a list, tuple, or dict, and thus give us more than one associated sub-value

# Exercise: Restaurant

1. Define a dictionary, `menu`, where the keys are strings and the values are integers. Each key-value pair represents one item on a restaurant's menu. (Item and price.)
2. Define `total` to be 0.
3. Ask the user, repeatedly, to order something from the menu.
    - If the user enters the empty string, then stop asking and print `total`.
    - If the user enters the name of something on the menu (i.e., a key in `menu`), then add the price to `total`, and print the item, its price, and the new `total`.
    - If the user enters the name of something *not* on the menu, then give them a gentle scolding.
4. Print `total`.

Example:

    Order: toast
    toast costs 5, total is 5
    Order: tea
    tea costs 4, total is 9
    Order: elephant
    sorry, we're out of elephant today!
    Order: [ENTER]
    Total is 9

https://practice.lernerpython.com/classroom/ce06963a8d/ex-20    

```python
# dict paradigm 1: we define it once, at the top of the program, and then use it in the program
# here, it's like a read-only database

menu = {'sandwich':15, 'tea':5, 'apple':2, 'cake':7}
total = 0

while True:
    order = input('Order: ').strip()
    
    if order == '':            # did I get an empty string?
        break                  # stop the loop

    if order in menu:          # is the user's input string one of the keys in our dict?
        price = menu[order]    # go to the menu dict, and grab the value associated with the key in order, and assign to price
        total += price         # add price to the total
        print(f'{order} costs {price}; total is now {total}')

    else:
        print(f'Sorry, we are out of {order} today.')

print(f'total = {total}')    
```

# Dictionaries are mutable!

What does this mean?

- We can replace existing values with new ones
- We can add new items (i.e., key-value pairs) to the dict
- We can remove existing key-value pairs from the dict

```python
# we can replace an existing value with a new one

d = {'a':10, 'b':20, 'c':30}

# I want to replace the value for 'b' with 999
d['b'] = 999

d
```

```python
# what about adding a new key-value pair to a dict?
# it's not like a list, where we have a special "list.append" method
# rather, we just assign -- exactly as we did when replacing a value -- but we assign to a key that doesn't exist (yet)

d['x'] = 888   # because 'x' doesn't yet exist, it is created for us

d
```

```python
# what about removing a key-value pair?
# (this is pretty rare, in my experience)
# use the dict.pop method, which takes a key -- it returns the value that was removed

d.pop('x')   # this removes the key-value pair with a key of 'x', and returns the associated value
```

# Dict paradigm 2: Update values, keep keys stable

In this way of using dicts, we define a dict at the start of a program. The keys will remain the same (none added, none removed), but the values are initially set to be starter values. Over the course of the program, these values change to reflect what we've seen.



```python
# odds and evens

counts = {'odds':0, 
          'evens':0}

text = input('Enter numbers, separated by spaces: ').strip()

for one_word in text.split():
    if one_word.isdecimal():     # can we turn this string into an int?
        n = int(one_word)        # get an int from it, and assign to n

        if n % 2 == 1:           # when we divide by 2, is there a remainder of 1?
            print(f'\tAdding {n} to odds')
            counts['odds'] += 1  # add 1 to the odd count
        else:
            print(f'\tAdding {n} to evens')
            counts['evens'] += 1 # add 1 to the even count

    else:
        print(f'{one_word} is not numeric; ignoring')

print(counts)   # print what we have            
```

# Exercise: Vowels, digits, and others (dict edition)

1. Define a dict with three keys (`vowels`, `digits`, and `others`), where all values are 0.
2. Ask the user to enter a string.
3. Go through the string, one character at a time:
    - If it's a vowel (a, e, i, o, u), then add 1 to `vowels`.
    - If it's a digit (0-9), then add 1 to `digits`.
    - If it's neither, then add 1 to `others`.
4, Print the resulting dict.

https://practice.lernerpython.com/classroom/ce06963a8d/ex-21

```python
counts = {'vowels': 0,
          'digits': 0,
          'others': 0}

text = input('Enter text: ').strip()

for one_character in text:          # go through the string, one character at a time
    if one_character in 'aeiou':    # is it a vowel?
        print(f'\t{one_character} is a vowel')
        counts['vowels'] += 1       #    add 1 to the value of 'vowels'
    elif one_character.isdecimal(): # is it a decimal number?
        print(f'\t{one_character} is a digit')
        counts['digits'] += 1       #    add 1 to the value of 'digits'
    else:
        print(f'\t{one_character} is other')
        counts['others'] += 1       # otherwise, add 1 to the count of others

print(counts)        
```

# Dict values can be *anything*

What if, instead of integers, we have lists as our dict values?

```python
# odds and evens, list edition

# this dict will have strings for keys and *lists* as values
# instead of counting the odds and evens, we'll actually store them!

counts = {'odds':[], 
          'evens':[]}

text = input('Enter numbers, separated by spaces: ').strip()

for one_word in text.split():
    if one_word.isdecimal():     # can we turn this string into an int?
        n = int(one_word)        # get an int from it, and assign to n

        if n % 2 == 1:           # when we divide by 2, is there a remainder of 1?
            print(f'\tAdding {n} to odds')
            counts['odds'].append(n)     # append n to the end of the list
        else:
            print(f'\tAdding {n} to evens')
            counts['evens'].append(n)    # append n to the end of the list

    else:
        print(f'{one_word} is not numeric; ignoring')

print(counts)   # print what we have            
```

```python
# how can I retrieve the list associated with 'odds'?

counts['odds']
```

```python
# how can I get the first element of the list associated with 'odds'?
counts['odds'][0]
```

Paradigm #3 for working with dicts is: Start with an empty dictionary:
- As the program goes on, you'll add new key-value pairs -- keys based on the user's input, and values starting with 0, `[]`.
