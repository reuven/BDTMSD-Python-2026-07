# What is going on with these data structures?

In theory, you could write every program you would want using just 1s and 0s. We build these higher-level data structures so that we can think at a more human level about what we're doing.

So far, we've talked about a bunch of relatively simple data structures. 

- Integers -- whole numbers. We use these for counting, but we also use them for indexing.
- Floats -- numbers with a decimal point. We use these for accurate calculations.
- Strings -- text. Both for storing, reading, displaying. We also saw that we can use it for storing text and searching in it.
- Lists -- Any time you have a bunch of things
    - Filenames in a directory
    - People in an organization
    - IP addresses on a network
- Tuples -- you can think of them as lists, but when you have a variety of types of values you need to store. We think of these as "records"
    - Person -- first name, last name, shoe size
    - Employee -- ID, name, department
    - Products -- ID, name, division, sales last year

We could have a more-than-adequate programming language just with these types. They're already more and higher level than the computer has to deal with!

Today, we're going to start talking about dictionaries, aka "dicts." They are versatile, flexible, and super fast. Not only are many programming problems easier to solve when we use dicts, but Python itself uses dicts behind the scenes in many places.

In many ways, the key to programming is to put your data into a format / structure that makes it easy to work with. Then your program becomes much simpler.



# Dictionaries

Python calls them dicts, but they exist in other languages, too:

- Hashes
- Hash maps
- Maps
- Hash tables
- Key-value pairs
- Name-value pairs
- Associative arrays

The basic idea of a dict is that you don't store one thing, as with a list. Rather, you store *two* things, the key and the value. The key comes in place of the index in a list.

In a list, you can store whatever you want, but the index is determined by when you added the value to the list. The index starts with 0, then 1, then 2, until `len(the_list) - 1`.

In a dict, you can determine what the key (the index, basically) is! If you want to use people's ID numbers, you can do that. They don't need to be sequential. Moreover, they don't need to be integers.

# Basic rules and syntax for dicts

1. We create a dict with `{}`.
2. Each key-value pair has a `:` between the key and the value.
3. Pairs are separated from one another with `,`, as in a list or tuple.
