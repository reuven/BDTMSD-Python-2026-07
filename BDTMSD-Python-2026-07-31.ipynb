# Loop controls

We've seen that a `for` loop's body can contain any code whatsoever. It can contain:

- `print`
- `input`
- `for` loop ("nested" loop)
- `if` statements

But so far, the entirety of a loop has executed each time. (Unless you use `if` to cordon off which part of a loop body runs at various points.)

- What happens if we know that we don't need to execute the loop body this time around?
- What happens if we know that we have already completed the mission that the loop had to do -- can we avoid all of the remaining iterations?

We have two statements in Python that we can for these purposes:

- `break` means: Stop the loop now. Exit from the loop body, and continue executing the program immediately after the loop body. If there is nothing after, then the program ends. This is almost always inside of an `if` block, because we only want to stop the loop in certain cases. This is used when we have accomplished our mission, and don't need/want to waste time on additional iterations that have no chance of helping.
- `continue` means: Jump immediately to the next iteration. Don't execute the rest of the loop body. If there is no next iteration, then the loop ends. This is also almost always used inside of an `if` block.  Also, this is done when you don't want to end the loop, but the current iteration is irrelevant or unnecessary, and there's no reason to spend time even checking/thinking about it.

It's pretty common for the top of a loop body to have a bunch of `if` statements, checking various values and statuses. if the condition is `True`, then we can either immediately stop the loop or just stop this iteration. We do this so that the rest of the loop body can be simpler, and not indented.

```python
# print all of the letters
s = 'abcde'

for one_letter in s:
    print(one_letter)
```

```python
# print all of the letters, except for look_for
s = 'abcde'
look_for = 'd'

for one_letter in s:
    if one_letter == look_for:    # go onto the next iteration if we're on 'd'
        continue
    
    print(one_letter)
```

```python
# what if we use break instead of continue?

s = 'abcde'
look_for = 'd'

print('Before')
for one_letter in s:
    if one_letter == look_for:  
        break
    
    print(one_letter)
    
print('After')    
```

# When would we use `break` or `continue`?

- If we're looking through a file, and we get blank lines, we can `continue` onto the next line
- If we're looking through a database, and we get employees who are no longer working at the company. We can `continue` onto the next one.
- If we're looking through a budget, and it goes over the annual income, we can `break` from the loop -- no point in continuing.

# A small warning:

The `break` statement exits the loop right away. This statement has been in Python from the beginning, 30+ years.

A few years ago, the core developers added a new statement, `breakpoint`, which is helpful in debugging Python programs. You don't want to mix these up! And if you have auto-complete on, be extra careful!

# Exercise: Sum digits

1. Define `total` to be 0.
2. Ask the user to enter a string containing digits.
3. Go through each character in the string.
    - If it's a digit, add it to `total`
    - If it's a `.`, stop summing and print the `total` and end.
    - If it's not a digit and not a `.`, then scold the user.

Example:

    Digits: 27ab.1
    got 2, total is 2
    got 7, total is 9
    got a -- not numeric; try again
    got b -- not number; try again
    got . -- stopping here
    Total is 9

https://practice.lernerpython.com/classroom/ce06963a8d/ex-11

```python
total = 0

numbers = input('Enter numbers: ').strip()

for one_digit in numbers:
    if one_digit == '.':   # check -- did we get a .?
        print(f'Got . -- stopping here')
        break
        
    if not one_digit.isdecimal():   # check -- did we get a non-decimal digit?
        print(f'{one_digit} is not numeric!')
        continue

    # one_digit must be a decimal digit, which we can convert, add, etc.
    n = int(one_digit)
    total += n
    print(f'Added {n}; total is {total}')

print(total)
```

# This isn't enough

`for` loops are extremely common in Python, in part because so many data structures (not just strings!) know how to work. 

But the idea of a `for` loop is that you start, and it goes through all of the elements, and then it stops. You assume that by going through each element, you'll get what you want.

But there are times when we don't know how many iterations we'll need -- not going through all of the characters in a string, or even all of the numbers in a range. We just don't know if it'll take a long time (many iterations) or a short time (few iterations).

For such cases, it's important that we know what the condition is for stopping. I don't know how many iterations I'll need, but I know when I've gotten to the end.

For such cases, we instead use the `while` loop.

You can think of `while` as `if`, but with repetition:

- `while`, like `if`, needs a condition to its right. It looks to the right and needs either `True` or `False`.
- If the condition is `True`, then the loop body executes. (Notice that there isn't a loop variable, as with `for`.)
- When the loop body finishes, Python goes back to the `while` and evaluates the condition *again*. If it's `True`, then the loop body runs again.
- So long as the condition is `True`, the body executes.
- When the condition is `False`, the program continues right after the loop body.
- Almost certainly, you will want your loop body to assign to or change one or more variables used in the `while` loop's condition. If you don't do that, then you might well find yourself with an infinite loop.

```python
x = 5

print('Before')
while x > 0:   # this is only checked at the top of each iteration, not in the middle
    print(x)
    x = x - 1   # decrement x!
print('After')    
```

```python
# a semi-common while loop

while True:     # this is definitely an infinite loop!
    name = input('Enter your name: ').strip()

    if name == '':  # this is our escape hatch -- if the user gives an empty string, we exit the loop
        break

    print(f'Hello, {name}!')
```

# Exercise: Sum to 100

1. Set `total` to 0
2. Ask the user, repeatedly, to enter a number.
    - If it's invalid, then scold the user.
    - If it's the empty string, then stop the program early
    - Convert the number to an int and add to `total`
3. When `total` is >= 100, stop the program and print `total`.

Example:

    Enter number: 20
    total is 20
    Enter number: 50
    total is 70
    Enter number: hello
    hello is not numeric
    Enter number: 29
    total is 99
    Enter number: 5
    total is 104

https://practice.lernerpython.com/classroom/ce06963a8d/ex-12

```python
total = 0


while total < 100:
    s = input('Enter number: ').strip()

    if s == '':
        print(f'Empty string; exiting')
        break
    
    if not s.isdecimal():
        print(f'{s} is not numeric; try again')
        continue   # we can use it with a while, too!

    n = int(s) 
    total += n
    print(f'Added {n}; total is now {total}')

print(total)
```
