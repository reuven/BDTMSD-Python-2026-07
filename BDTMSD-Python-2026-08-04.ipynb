# Lists are mutable

You might remember that strings are *immutable*, they cannot be changed. This doesn't mean that the association between a variable and a string value are permanent; that's a different idea, that of a *constant*, which Python doesn't have. You can always re-assign a variable's value.

But can you change the value itself? In the case of a string, the answer is "no."

```python
text = 'abcdefghij'

# if I want to change the string, maybe I can assign to text[0]

text[0] = '?'
```

```python
text[:3]  # this returns a new string, based on text, but does not modify text
```

```python
mylist = [10, 20, 30, 40, 50]

# we can change the list!
```

# Be careful when assigning!

1. If we say `mylist = [10, 20, 30]`, we are assigning a value to a variable. If the variable didn't exist before, it does now.
2. If we say `mylist[0] = 40`, now we are not assigning a value to a variable. We are *mutating* an existing value that the variable refers to. This means that when we mutate a list through assignment, the index must exist!

```python
mylist[98654] = 999
```

```python
mylist
```

```python
mylist[1] = 40
```

```python
mylist
```

# Who cares?

Why do we care about mutation so much?

1. Mutable values are closer to the real world -- we can add, modify, and remove items.
2. Two variables can refer to the same value. If the value changes, that is reflected in both variables. This can be a real pain to figure out and debug!

```python
mylist = [10, 20, 30]  # these two variables both refer to the same list! Modifying one also modifies the other
x = mylist

mylist[0] = '!'   # changed the value at mylist[0]
x[1] = '?'        # changed the value at x[1]
```

```python
# what does mylist look like?
mylist
```

```python
x
```

# Exercise: Vowels, digits, and others (list edition)

1. Ask the user to enter some text.
2. We'll also define three variables -- `vowels`, `digits`, and `others`, and assign them all `[]`, the empty list.
3. Go through each character in the text, one at a time:
    - If the character is a vowel (a, e, i, o, u), then append it to `vowels`
    - If the character is a digit (0-9), then append it to `digits`
    - Otherwise, append it to `others`
4. Print the variables -- the three lists.

https://practice.lernerpython.com/classroom/ce06963a8d/ex-14


```python
vowels = []
digits = []
others = []

text = input('Enter text: ').strip()

for one_character in text:
    if one_character in 'aeiou':       # is one_character in our string of vowels? 
        vowels.append(one_character)   #    we want to add one_character (a vowel) to the end of vowels (a list) -- we use list.append
    elif one_character.isdecimal():    # is one_character a decimal digit?
        digits.append(one_character)
    else:
        others.append(one_character)

print(f'vowels = {vowels}')        
print(f'digits = {digits}')
print(f'others = {others}')
```

# Exercise: Multiples of

1. Define three empty lists -- `mult2`, `mult3`, `mult5`
2. Ask the user to enter a number.
3. Iterate from 0 until (not including) that number, using `range`.
4. Check each number to see if it's divisible by 2, 3, and/or 5.
5. (Remember that the `%` operator, "modulo", returns the remainder from division. So `10 % 3` returns 1, because 10 / 3 is 3 with a remainder of 1. And we get the remaidner.
6. Append each number to one or more of these lists -- whichever are appropriate. If the number is 30, which is divisible by all 3, then all three lists should get that number.
7. When you're done, print all three lists.

https://practice.lernerpython.com/classroom/ce06963a8d/ex-15

```python
mult2 = []
mult3 = []
mult5 = []

maxnum = input('Enter max number: ').strip()
maxnum = int(maxnum)

for number in range(maxnum):

    if number % 2 == 0:         # if there is no remainder when dividing by 2, it's divisible by 2!
        mult2.append(number)    #   append it to the end of mult2

    if number % 3 == 0:       # if there is no remainder when dividing by 3, it's divisible by 3!
        mult3.append(number)

    if number % 5 == 0:       # if there is no remainder when dividing by 5, it's divisible by 5!
        mult5.append(number)

print(f'mult2 = {mult2}')        
print(f'mult3 = {mult3}')        
print(f'mult5 = {mult5}')        
```

```python
# what about this:

text = 'abcd:ef:ghi:jk'

# can I get a list of strings from this?

list(text)
```

```python
# what if I want to get a list of strings back from text, using : as the delimiter (separator)?
# meaning: I want to cut the text into pieces, chopping wherever I see :

# we can use the str.split method to do this, passing the delimiter as the argument

text.split(':')  # I'll always get a list of strings back
```

```python
text.split('e')
```

`str.split` is a very standard way to take information in a file, from the user, from a Web site, etc., and break it into pieces so that I can work on each piece. I get a list back, and then can pull out individual list elements or (even more common) iterate over the list, doing something with each piece.

```python
text = 'abcd,ef,ghi'   # CSV (comma-separated values) record

text.split(',')
```

```python
# what about a sentence?

text = 'hello to everyone out there'
text.split(' ')
```

```python
# what if the user isn't so good at the space bar.

text = 'hello  to    everyone out      there'
text.split(' ')  # this means: chop every time you see a ' ' character
```

```python
# the solution here is to invoke str.split with NO ARGUMENT AT ALL!
# that is taken as a sign that any/all whitespace should be taken as a single delimiter, no matter how many are in a row

text.split()   # no argument!
```

# Exercise: Sum numbers

1. Define `total` to be 0.
2. Ask the user to enter some numbers, separated by spaces.
3. Go through each number in the user's input:
    - If it cannot be turned into an int, scold the user and go on
    - Add the integer to `total`, printing the number and the running `total`.
4. Print the final `total`.

https://practice.lernerpython.com/classroom/ce06963a8d/ex-16

```python
total = 0

# text = input('Enter numbers: ').strip()   # this gives us a string that we still have to split into elements

numbers = input('Enter numbers: ').split()   # this puts a list of strings in "numbers"

for one_number in numbers:   # iterate over the list of strings

    if not one_number.isdecimal():
        print(f'\tIgnoring {one_number}, which is non-numeric')
        continue

    n = int(one_number)
    total += n

    print(f'added {n}, total is {total}')

print(f'Final total is {total}')
```
