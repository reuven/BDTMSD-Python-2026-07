# List refresher

A list is created with `[]` and commas between the elements. A list can contain any number of values, including zero. Traditionally, a list contains only items of one type, but that isn't enforced.

```python
mylist = [10, 20, 30, 40, 50, 60]

mylist[0]  # retrieves the first item
```

```python
mylist[1]  # retrieves the second item
```

```python
mylist[-1]  # retrieves the final item
```

```python
len(mylist)  # how many elements? This always gives us a number 1 more than the maximum index
```

```python
# slice

mylist[2:5]  # from 2 up to and not including 5
```

```python
mylist[:3]  # from the start until 3
```

```python
mylist[3:]  # from 3 until the end
```

```python
# for loops on lists!

for one_item in mylist:
    print(one_item)
```

# Lists are mutable

You might remember that strings are *immutable*, they cannot be changed. This doesn't mean that the association between a variable and a string value are permanent; that's a different idea, that of a *constant*, which Python doesn't have. You can always re-assign a variable's value.

But can you change the value itself? In the case of a string, the answer is "no."

```python
text = 'abcdefghij'

# if I want to change the string, maybe I can assign to text[0]

text[0] = '?'
```

```python
text[:3]  # this returns a new string, based on text, but does not modify text
```

```python
mylist = [10, 20, 30, 40, 50]

# we can change the list!
```
