# List refresher

A list is created with `[]` and commas between the elements. A list can contain any number of values, including zero. Traditionally, a list contains only items of one type, but that isn't enforced.

```python
mylist = [10, 20, 30, 40, 50, 60]

mylist[0]  # retrieves the first item
```

```python
mylist[1]  # retrieves the second item
```

```python
mylist[-1]  # retrieves the final item
```

```python
len(mylist)  # how many elements? This always gives us a number 1 more than the maximum index
```

```python
# slice

mylist[2:5]  # from 2 up to and not including 5
```

```python
mylist[:3]  # from the start until 3
```

```python
mylist[3:]  # from 3 until the end
```

```python
# for loops on lists!

for one_item in mylist:
    print(one_item)
```

# Lists are mutable

You might remember that strings are *immutable*, they cannot be changed. This doesn't mean that the association between a variable and a string value are permanent; that's a different idea, that of a *constant*, which Python doesn't have. You can always re-assign a variable's value.

But can you change the value itself? In the case of a string, the answer is "no."

```python
text = 'abcdefghij'

# if I want to change the string, maybe I can assign to text[0]

text[0] = '?'
```

```python
text[:3]  # this returns a new string, based on text, but does not modify text
```

```python
mylist = [10, 20, 30, 40, 50]

# we can change the list!
```

# Three ways we can modify a list

1. We can replace an existing value with a new one
2. We can add a new value to the list, making it longer
3. We can remove an existing value from the list, making it shorter



```python
# replace an existing value by assigning to a particular index

mylist
```

```python
mylist[0] = '!!!!'
```

```python
mylist
```

# Be careful when assigning!

1. If we say `mylist = [10, 20, 30]`, we are assigning a value to a variable. If the variable didn't exist before, it does now.
2. If we say `mylist[0] = 40`, now we are not assigning a value to a variable. We are *mutating* an existing value that the variable refers to. This means that when we mutate a list through assignment, the index must exist!

```python
mylist[98654] = 999
```

```python
mylist
```

```python
mylist[1] = 40
```

```python
mylist
```

```python
# we can also add a new value to our list, and make it longer
# (you can add values to a list wherever you want, but it's most common and most efficient to add to the end)

# add a new element to the end with the "list.append" method

mylist = [10, 20, 30]
mylist.append(40)   # the integer 40 has now been added to our list, as the final element

mylist
```

```python
len(mylist)
```

```python
# what can I add with append? whatever I want!

mylist.append('hello')
mylist
```

```python
mylist.append([100, 200, 300])
mylist
```

```python
len(mylist)
```

```python
# what if I want to add a bunch of things?
# what if I have a list, and I want to add all of its values to the end of another existing list?

# option 1: run a for loop, and append many times
mylist = [10, 20, 30]
newlist = [100, 200]

for one_item in newlist:
    mylist.append(one_item)  # add the item from newlist to the end of mylist

mylist    
```

```python
newlist
```

```python
# option 2: have Python do the hard work for us!
mylist = [10, 20, 30]
newlist = [100, 200]

mylist += newlist   # this runs a for loop on newlist, appending each element to mylist
mylist
```

```python
# we can remove existing values from our list
# (once again, we can remove them in a variety of ways; I'll show you the most common and efficient way.)

# the list.pop method returns the final value and removes it

mylist = [10, 20, 30, 40, 50]
mylist.pop()
```

# Who cares?

Why do we care about mutation so much?

1. Mutable values are closer to the real world -- we can add, modify, and remove items.
2. Two variables can refer to the same value. If the value changes, that is reflected in both variables. This can be a real pain to figure out and debug!

```python
mylist = [10, 20, 30]  # these two variables both refer to the same list! Modifying one also modifies the other
x = mylist

mylist[0] = '!'   # changed the value at mylist[0]
x[1] = '?'        # changed the value at x[1]
```

```python
# what does mylist look like?
mylist
```
