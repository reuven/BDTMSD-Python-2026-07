# DRY -- don't repeat yourself!

We talked about the DRY rule when we discussed loops a few weeks ago. The idea is: Why repeat the same code, if you want have a loop do it for you.

This has a bunch of advantages:

- You don't have to write as much
- You don't have to modify/maintain/debug as much (!!!!)
- It's easier to think about code in this way

We discussed the DRY rule when we talked about loops, saying that if you have the same code several times in a row, then you can use a loop to "DRY up" the code.

The same is true, more or less, with functions: If you have the same code in several places in your program, then you can put that code in one place, give a name (a function), and then refer to it each time.

Functions give us, in addition to DRYing up our code, a few other advantages:

1. They allow us to easily share code, and to help other people reduce how much code they need to write.
2. Abstraction -- the power of abstraction, meaning that we can think about the high-level design, and not worry about the low-level details.

A function gives one name to a large number of actions. We could do each of these actions ourselves, one by one, but it's easier to think about and communicate once we have them gathered into a single function.

Then, of course, once a function has been defined, we can use it inside of another function's definition, building higher and higher towers of abstraction.

# Functions are values, too

In many programming languages, there is a distinction between "data" and "functions" (or nouns and verbs, if you will). Not so in Python! In Python, a function is a data structure, and is a value.

In other words, our verbs are also nouns! When you define a function, you're create a "function object," a set of instructions -- blueprints to what you want to do. When you *run* the function, then those instructions are executed.

How do we distinguish between examining/working with the instructions rather than executing them? The difference is all in `()`. If you put `()` after a function name, it is invoked/called/executed. If you don't put the `()` there, then the function will not be called. Instead, you'll get a function object, i.e., the blueprints, which is almost certainly not what you want, at least when you're starting off.

```python
# example of function definitions vs. function execution

d = {'a':10, 'b':20, 'c':30}
```

```python
# this is just fine!

for key, value in d.items():  # here, we're iterating over the result of invoking d.items()
    print(f'{key}: {value}')
```

```python
# this is not...

for key, value in d.items:     # look, I forgot the ()! we're trying to iterate over the function definition itself... impossible
    print(f'{key}: {value}')
```

# Let's define a function!

If we want to define a function, we use the reserved word `def`:

- We start the line with `def`
- Then we give the name of the function. This is a variable assignment, so all rules and conventions for variable names apply to function names, too. Usually all lowercase letters with `_` between the words
- After the function name, we have `()`. This is where we will (not just yet) put the *parameters*, the variables into which arguments are assigned. For now, we have no parameters, and so it's empty `()`.
- Then we have `:`. This means the end of the line, and the next line will start a block -- the function body.
- In the function body, you can write as much (or as little) code as you want. Any Python code (with very rare exceptions) can be put in a function body.
- When you define your function, the body does *not* execute! You're merely defining the function, not running it.



```python
def hello():
    print('Hello out there!')
```

```python
hello()   # Invoke the function with ()
```

# Exercise: Calculator

1. Write a function, `calc`, that when we run it does the following:
    - It asks the user to enter a first number
    - It asks the user to enter an operator, either `+` or `-`
    - It asks the user to enter a second number
2. For now, we'll assume that the user gave us inputs that only contain digits.
3. Print the full expression, plus the result.
4. If the operator isn't known, the result can be "Invalid" or "unknown."

Example:

    First: 10
    Operator: +

```python
def calc():
    first = input('First: ').strip()
    op = input('Operator: ').strip()
    second = input('Second: ').strip()

    first = int(first)
    second = int(second)

    if op == '+':
        result = first + second
    elif op == '-':
        result = first - second
    else:
        result = f'(Bad operator {op})'

    print(f'{first} {op} {second} = {result}')
```

# Arguments vs. parameters

Most programmers I've met confuse these two words:

- An argument is a value that you pass to a function when you invoke it.
- A parameter is a variable to which a value is assigned.

A parameter is a special kind of variable: We (almost) never assign to it! Rather, Python, as part of the mechanism of invoking a function and passing the arguments, makes that assignment for us.

```python
hello('world')   # Python, behind the scenes, assigns 'world' to first_name
```

```python
hello('Reuven')
```

```python
def hello(first_name, last_name): 
    print(f'Hello, {first_name} {last_name}!')
```

```python
hello('Reuven', 'Lerner')
```

```python
# what about our earlier versions of hello?

hello()  # no arguments
```

```python
hello('Reuven')  # 1 argument
```

Python doesn't let you have more than one version of a function at a given time. Unlike many other languages, where you can define a function several times, and the correct/appropriate one will work, Python gives you one shot at defining a function. If you define a function a second time, the second definition is considered authoritative.

This has to be, because `def` is assigning to the function name as a variable. Just as:

    x = 5
    x = 7   # at this point, does x remember it was ever 5? No!

If you define a function twice, the second time wins. If that means changing the number of parameters, then so be it.

```python
def hello(first_name, last_name): 
    print(f'Hello, {first_name} {last_name}!')
```

```python
# parameters:  first_name    last_name
# arguments:      'a'           'b'

hello('a', 'b')
```

# Return values

When we call `input`, we expect that it will (a) pause the program and (b) ask the user for input and (c) give us a string back.

This "give us something back" is known as a "return value." Every function in Python returns a value. If we don't specify what a function returns, then it returns the special value `None`.

Our `hello` (and `calc`) functions both print their output on the screen. That is very different from returning a value:

- The return value from a function can be assigned to a variable. Something printed on the screen cannot.
- The return value from a function can be used as the input to another function. Something printed on the screen cannot.
- Generally speaking, it's considered far better to return a value, and let the caller decide what to do, rather than print on the screen, and force the user's hand.

We've seen return values:

- `input` returns a string
- `len` returns an integer
- `str.strip` returns a string
- `str.split` returns a list of strings

We can return a value from our Python function with the `return` keyword. It's not a function! To its right, put the value that you want to return to the caller.

If you want, you can just say `return`, and that stops the function's execution and returns `None`.

When a function returns, its execution stops right away.

If you don't have a `return` in your function, then it'll just execute all of the way through.

It's very common to have `return` inside of an `if` statement. That allows you to return something only if something else is true.

If you're using Jupyter, then the difference between printing and returning is even harder to understand -- that's because Jupyter displays a function's return value if it isn't assigned to something else.
