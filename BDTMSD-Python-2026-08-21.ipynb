# Where are we?

So far, we've seen two parts of Python (and programming in general):

- Basic syntax, including:
    - assignment and values
    - conditional execution with `if`
- Data structures (our nouns)
    - booleans (`True` and `False`)
    - integers and floats
    - strings
    - lists
    - tuples
    - dicts
    - files

We now have a decent collection of data structures, which will allow us to organize our information, and ways that we can work with them, including a number of built-in functions and methods -- the verbs in a programming language.

Technically speaking, we don't need anything more. With just these tools, we can build any program we want.

However, you really don't want to stop here. You really do want functions.

# What do functions give us?

Functions are the verbs in a programming language. They do things:

- `print`
- `input`
- `sum`
- `type`
- `len`
- `int`
- `str`
