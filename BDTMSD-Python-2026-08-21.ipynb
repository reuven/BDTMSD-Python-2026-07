# DRY -- don't repeat yourself!

We talked about the DRY rule when we discussed loops a few weeks ago. The idea is: Why repeat the same code, if you want have a loop do it for you.

This has a bunch of advantages:

- You don't have to write as much
- You don't have to modify/maintain/debug as much (!!!!)
- It's easier to think about code in this way

We discussed the DRY rule when we talked about loops, saying that if you have the same code several times in a row, then you can use a loop to "DRY up" the code.

The same is true, more or less, with functions: If you have the same code in several places in your program, then you can put that code in one place, give a name (a function), and then refer to it each time.

Functions give us, in addition to DRYing up our code, a few other advantages:

1. They allow us to easily share code, and to help other people reduce how much code they need to write.
2. Abstraction -- the power of abstraction, meaning that we can think about the high-level design, and not worry about the low-level details.

A function gives one name to a large number of actions. We could do each of these actions ourselves, one by one, but it's easier to think about and communicate once we have them gathered into a single function.

Then, of course, once a function has been defined, we can use it inside of another function's definition, building higher and higher towers of abstraction.

# Functions are values, too

In many programming languages, there is a distinction between "data" and "functions" (or nouns and verbs, if you will). Not so in Python! In Python, a function is a data structure, and is a value.

In other words, our verbs are also nouns! When you define a function, you're create a "function object," a set of instructions -- blueprints to what you want to do. When you *run* the function, then those instructions are executed.

How do we distinguish between examining/working with the instructions rather than executing them? The difference is all in `()`. If you put `()` after a function name, it is invoked/called/executed. If you don't put the `()` there, then the function will not be called. Instead, you'll get a function object, i.e., the blueprints, which is almost certainly not what you want, at least when you're starting off.

```python
# example of function definitions vs. function execution

d = {'a':10, 'b':20, 'c':30}
```

```python
# this is just fine!

for key, value in d.items():  # here, we're iterating over the result of invoking d.items()
    print(f'{key}: {value}')
```

```python
# this is not...

for key, value in d.items:     # look, I forgot the ()! we're trying to iterate over the function definition itself... impossible
    print(f'{key}: {value}')
```

# Let's define a function!

If we want to define a function, we use the reserved word `def`:

- We start the line with `def`
- Then we give the name of the function. This is a variable assignment, so all rules and conventions for variable names apply to function names, too. Usually all lowercase letters with `_` between the words
- After the function name, we have `()`. This is where we will (not just yet) put the *parameters*, the variables into which arguments are assigned. For now, we have no parameters, and so it's empty `()`.
- Then we have `:`. This means the end of the line, and the next line will start a block -- the function body.
- In the function body, you can write as much (or as little) code as you want. Any Python code (with very rare exceptions) can be put in a function body.
- When you define your function, the body does *not* execute! You're merely defining the function, not running it.



```python
def hello():
    print('Hello out there!')
```

```python
hello()   # Invoke the function with ()
```
