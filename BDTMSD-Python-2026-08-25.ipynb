Let's say that we only want people to call `add` with two numbers (either `int` or `float`). How can we do that?

You might think that a good place to put such notes would be in the comments of the function. But there are two problems with this:

1. The comments are not usually something people look at.
2. Comments are meant for the people who are writing and maintaining a function, not those who are using it.

In order to separate comments (for maintainers) from documentation (for users/callers of a function), Python uses what we call "docstrings."

The idea is that the first line of a function is a string. Not an assignment to a string, and not printing a string, but just a string. If Python sees that when you have defined your function, it takes that string to be a "docstring," meant for and displayed to the people who are using your function.

```python
def add(first, second):
    'This is the best function in the world!'
    return first + second
```

```python
add(2, 5)
```

Docstrings don't affect the actual implementation or execution of a function. They are simply there to be used when someone wants to inspect a function:

- In Jupyter, you can invoke the "help" function, and find the docstring
- In an IDE, you can hover over the function name, and you'll typically see the docstring
- Sometimes, it's extracted onto Web sites

```python
help(add)   # I'm passing the function `add` (not executing it!) to `help` as an argument
```

# API documentation -- documentation for people who will invoke your function -- should include:

- Expects - what does the function expect to receive as arguments? 
- Modifies - what will the function modify? (Could be arguments, could be files, could be a database...)
- Returns - what does the function return -- both type and semantics

It's typical and traditional to use a triple-quoted string for your docstring.

Also: The first line is usually seen as special by Python, as a sort of summary/synopsis of what the function does.

```python
def add(first, second):
    """
    Returns the sum of two numbers.

    Expects: Two arguments, both integers or floats. Passing other types might result in an error.
    Modifies: Nothing
    Returns: The sum of the two input arguments.
    """

    return first + second
```

```python
add(2, 9)
```

```python
help(add)
```
