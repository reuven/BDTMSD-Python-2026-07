Let's say that we only want people to call `add` with two numbers (either `int` or `float`). How can we do that?

You might think that a good place to put such notes would be in the comments of the function. But there are two problems with this:

1. The comments are not usually something people look at.
2. Comments are meant for the people who are writing and maintaining a function, not those who are using it.

In order to separate comments (for maintainers) from documentation (for users/callers of a function), Python uses what we call "docstrings."

The idea is that the first line of a function is a string. Not an assignment to a string, and not printing a string, but just a string. If Python sees that when you have defined your function, it takes that string to be a "docstring," meant for and displayed to the people who are using your function.

```python
def add(first, second):
    'This is the best function in the world!'
    return first + second
```

```python
add(2, 5)
```

Docstrings don't affect the actual implementation or execution of a function. They are simply there to be used when someone wants to inspect a function:

- In Jupyter, you can invoke the "help" function, and find the docstring
- In an IDE, you can hover over the function name, and you'll typically see the docstring
- Sometimes, it's extracted onto Web sites

```python
help(add)   # I'm passing the function `add` (not executing it!) to `help` as an argument
```

# API documentation -- documentation for people who will invoke your function -- should include:

- Expects - what does the function expect to receive as arguments? 
- Modifies - what will the function modify? (Could be arguments, could be files, could be a database...)
- Returns - what does the function return -- both type and semantics

It's typical and traditional to use a triple-quoted string for your docstring.

Also: The first line is usually seen as special by Python, as a sort of summary/synopsis of what the function does.

```python
def add(first, second):
    """
    Returns the sum of two numbers.

    Expects: Two arguments, both integers or floats. Passing other types might result in an error.
    Modifies: Nothing
    Returns: The sum of the two input arguments.
    """

    return first + second
```

```python
add(2, 9)
```

```python
help(add)
```

```python
def calc(first, op, second):
    """
    Perform a simple math operation, and return a string.

    Expects: Three arguments
        - first, an int or float
        - op, a string with either '+' or '-' -- others won't cause errors, but they won't help or get a result
        - second, another int or float
    Modifies: Nothing
    Returns: A string with first, op, second, and the result, as in '2 + 3 = 5'.
    """
    if op == '+':
        result = first + second
    elif op == '-':
        result = first - second
    else:
        result = f'(Unknown operator {op})'

    return f'{first} {op} {second} = {result}'

calc(10, '+', 3)    
```

# Keyword arguments

So far, when we have invoked functions, we have done so with *positional arguments*, meaning that the first argument is assigned to the first parameter, the second to the second, etc. If there were 10 parameters in our function, we would have to pass 10 arguments, and there would be no flexibility on the order.

Keyword arguments give us that kind of flexibility.

The idea is that instead of passing arguments as just values, we pass them in `name=value` style, dictating to Python what argument should be assigned to which parameter. This can be the same as the positional ordering, or different.

There are some cases in which a parameter will only allow positional values, and others were it will only allow keyword values. We're ignoring that for now.

```python
def sub(first, second):
    return first - second
```

```python
# positional

# parameters: first  second
# arguments:     10     3

sub(10, 3)
```

```python
# keyword arguments
# they must all be in the form of name=value

# parameters:   first     second
# arguments:     10          3

sub(first=10, second=3)
```

```python
# one of the nice things about keyword arguments is that I can mix them up

# parameters:    first    second
# arguments:       10       3

sub(second=3, first=10)
```

```python
# can we mix and match positional and keyword arguments?
# yes, so long as all of the positionals come first
```

```python
sub(10, second=3)   # no problem
```

```python
sub(first=10, 3)     
```

```python
# you can get into other problems, too:

sub(3, first=10)   # what's going on here?
```

Generally speaking, you can choose whether you pass arguments positionally or as keywords. I would say that the more parameters there are for a function, the more likely you'll want to use keyword arguments, just to keep things clearer and cleaner.

```python
def calc(first, op, second):
    """
    Perform a simple math operation, and return a string.

    Expects: Three arguments
        - first, an int or float
        - op, a string with either '+' or '-' -- others won't cause errors, but they won't help or get a result
        - second, another int or float
    Modifies: Nothing
    Returns: A string with first, op, second, and the result, as in '2 + 3 = 5'.
    """
    if op == '+':
        result = first + second
    elif op == '-':
        result = first - second
    else:
        result = f'(Unknown operator {op})'

    return f'{first} {op} {second} = {result}'

```

# Default argument values

We've seen a number of examples of functions (and methods) where there were obviously some default values. That is, we could call them with an argument... or we could call them without an argument, and they still worked.

With the tools we have so far, we cannot do that ourselves.

```python
def add(first, second):
    return first + second

add(10, 2)    
```

```python
add(10)
```

The way that we can give a parameter a default argument value looks identical to how we pass keyword arguments! It's in the style of `name=value`. 

The big difference(s) are:

- This is in the function definition, not when we invoke it
- This gives us flexibility when we invoke the function

Because of Python's syntax rules, just as all positional arguments need to come before all keyword arguments, all mandatory parameters need to be before all optional parameters. That is, if you have any default argument values, they need to be on the *last* of the parameters, and not before.

```python
def add(first, second=10):    # in this version, the parameter "second" has a default argument value of 10
    return first + second

```

```python
# parameters:   first    second
# arguments:     10        2

add(10, 2)    
```

```python
# parameters:   first    second
# arguments:     10        10

add(10)
```

The function itself, when it runs, is blind to the assignment of arguments to parameteters. The function doesn't know whether the invoker used positional or keyword arguments, or if arguments weren't passed, and then default values were used.

How many parameters can have default argument values? As many as you want, but they all have to be named after parameters without defaults.

In other words, mandatory parameters come before optional parameters.

If you have multiple optional parameters, then you can pass arguments to some of them, and not to others. The way to do that is with keyword arguments, which "skip over" the parameters with defaults you don't want to assign to.
