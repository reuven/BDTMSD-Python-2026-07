# Review of modules

- A module is a collection of definitions -- variables (data), functions, data types (classes)
- When you say `import MODNAME`, Python looks for `MODNAME.py` in its module search path, which is in `sys.path` -- a list of strings
- After importing `MODNAME`, you have a variable `MODNAME` defined, which is a module object -- basically a warehouse for names, all of the names that were defined in the module file, `MODNAME.py`
- The names that are defined in your module are all attributes on that module -- that is, the module, followed by `.`, followed by the attribute name.
- The whole point of importing a module is to get these definitions, and have access to them.
- If you want to use a particular name from a module a great deal, then you might want to have direct access to it -- not say `MODNAME.x`, but rather just say `x`. We saw that you can say `from MODNAME import x`, and then it's defined as a global variable in your program.
