# Review of modules

- A module is a collection of definitions -- variables (data), functions, data types (classes)
- When you say `import MODNAME`, Python looks for `MODNAME.py` in its module search path, which is in `sys.path` -- a list of strings
- After importing `MODNAME`, you have a variable `MODNAME` defined, which is a module object -- basically a warehouse for names, all of the names that were defined in the module file, `MODNAME.py`
- The names that are defined in your module are all attributes on that module -- that is, the module, followed by `.`, followed by the attribute name.
- The whole point of importing a module is to get these definitions, and have access to them.
- If you want to use a particular name from a module a great deal, then you might want to have direct access to it -- not say `MODNAME.x`, but rather just say `x`. We saw that you can say `from MODNAME import x`, and then it's defined as a global variable in your program.

# How can we write our own modules?

- A module is a text file containing Python code.
- The module must have a `.py` extension
- It must be in `sys.path` -- the easiest thing for us, right now, is to have it in the same directory as your program
- In the module file, just define variables and functions

# What happens when you `import` a module?

1. Python searches for the module file
2. Python loads the module into memory   <---- what does this mean?
    - Python executes the module file
    - It runs, line by line, inside of Python
    - That's how variables are defined. That's how functions are defined.
4. You can use the module

The fact that a module is executed when it is imported means that you can have *any* code you want inside of the module. **PLEASE DO NOT DO THIS!** You could have `print` and other statements in the module... but it's frowned upon.

# Once you `import` a module, it's around for the duration of your Python program

Normally, this isn't much of an issue: You run your program, the module is imported, and that's fine until the program ends.

But what if you modify the module while the program is still running? It doesn't matter -- the updates will be ignored. 

One way to force a module to be reloaded (which, I stress, is rarely needed in normal programming, even if it's important when using Jupyter) is to use the `importlib` module that comes with Python:

```python
from importlib import reload
```

Then we I can say `reload(mymod)`, and it'll be force-reloaded from disk.

# Lazy loading

Because a module executes when you `import` it, this can mean that it takes a long time for your program to start up. 

In Python 3.15, which is coming out next month, you'll be able to say

```python
lazy import mymod   # this means: wait to load the module until you first use a value in it
```
