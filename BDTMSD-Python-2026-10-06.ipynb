# Joins

The basic idea behind joining is that you have two different data frames that share an index -- or that share values in an index -- and you want to combine these into a single data frame.

This is very common if you have a large data set that is broken up into multiple files. Typically, each file (i.e., data frame) will have a number of different columns. So a company might have one data set with employee names and personal information, and another one with salary information, and yet another with performance information. If we want to do some analysis that uses information from all three of these data sets, then we'll need (want) to combine them into a single Pandas data frame, and then we can perform our query.

Joining would mean that all three of these data frames share an index, or at least share an index's values. You join across the index(es), get a single, large (wide) data frame, and then perform your queries.

If you don't want to join on the index, but want to use a non-index column, then you can use what Pandas calls "merge." Merging is basically the same as joining, except that it can be done on any column.

You'll often find that you'll need to use `set_index` to move a column into the index (if you want to `join`), or `reset_index` to move a column *out* of the index (if you want to `merge`).

```python
import pandas as pd
from pandas import Series, DataFrame
```

```python
oil_prices = DataFrame([[1, 60],
                        [2, 65],
                        [3, 63],
                        [4, 69],
                        [5, 72],
                        [6, 65],
                        [7, 60],
                        [8, 55],
                        [9, 60],
                        [10, 62],
                        [11, 63],
                        [12, 65]],
                      columns=['month', 'price'])
oil_prices
```

```python
# I want to know the (mean) price of oil, for each month... that is already in my oil_prices data frame, right?
# Except that I want to see the names of the months, not the numbers.

months = DataFrame([[1, 'January'],
                     [2, 'February'],
                     [3, 'March'],
                     [4, 'April'],
                     [5, 'May'],
                     [6, 'June'],
                     [7, 'July'],
                     [8, 'August'],
                     [9, 'September'],
                     [10, 'October'],
                    [11, 'November'],
                     [12, 'December']],
                  columns=['number', 'name'])

months
```

```python
oil_prices = oil_prices.set_index('month')
months = months.set_index('number')
```

```python
oil_prices
```

```python
months
```

```python
# in this join,

# oil_prices is known as the "left side"
# and months is known as the "right side"

oil_prices.join(months)
```

When you invoke `join`, the index on the left-hand data frame determines what will be included. If your left hand side's index has more values than the right-hand side index, then you'll have NaN in your result. But if the left side is a subset of the right side, that won't happen.

# Exercise: Joining

1. Create a data frame with the names of people in your family, and their ages.
2. Create a second, separate data frame with the names and shoe sizes of people in your family.
3. Join these together, and find the mean age per shoe size.

https://practice.lernerpython.com/classroom/ce06963a8d/ex-65

```python
ages = DataFrame([['Reuven', 56],
                  ['SF', 54],
                  ['AMLF', 25],
                  ['SBLF', 23],
                  ['ADLF', 20]],
                columns=['name', 'age'])

ages
```

```python
shoe_sizes = DataFrame([['Reuven', 46],
                  ['SF', 40],
                  ['AMLF', 40],
                  ['SBLF', 40],
                  ['ADLF', 44]],
                columns=['name', 'shoe_size'])

shoe_sizes
```

```python
(
    ages
    .set_index('name')
    .join(shoe_sizes
        .set_index('name'))
    .groupby('shoe_size')['age'].mean()
)
```

# What about `merge`?

The idea of `merge` is that you can join on any columns, not just on the index.

```python
(
    ages
    .merge(shoe_sizes)
)
```

```python
help(ages.merge)
```

```python
ages.columns
```

```python
shoe_sizes.columns
```

```python
ages.columns.intersection(shoe_sizes.columns)
```

```python
(
    ages
    .merge(shoe_sizes, on='name')   # make the join column explicit
)
```
