# Joins

The basic idea behind joining is that you have two different data frames that share an index -- or that share values in an index -- and you want to combine these into a single data frame.

This is very common if you have a large data set that is broken up into multiple files. Typically, each file (i.e., data frame) will have a number of different columns. So a company might have one data set with employee names and personal information, and another one with salary information, and yet another with performance information. If we want to do some analysis that uses information from all three of these data sets, then we'll need (want) to combine them into a single Pandas data frame, and then we can perform our query.

Joining would mean that all three of these data frames share an index, or at least share an index's values. You join across the index(es), get a single, large (wide) data frame, and then perform your queries.

If you don't want to join on the index, but want to use a non-index column, then you can use what Pandas calls "merge." Merging is basically the same as joining, except that it can be done on any column.

You'll often find that you'll need to use `set_index` to move a column into the index (if you want to `join`), or `reset_index` to move a column *out* of the index (if you want to `merge`).

# Exercise: Joining

1. Create a data frame with the names of people in your family, and their ages.
2. Create a second, separate data frame with the names and shoe sizes of people in your family.
3. Join these together, and find the mean age per shoe size.

https://practice.lernerpython.com/classroom/ce06963a8d/ex-65

```python
ages = DataFrame([['Reuven', 56],
                  ['SF', 54],
                  ['AMLF', 25],
                  ['SBLF', 23],
                  ['ADLF', 20]],
                columns=['name', 'age'])

ages
```

```python
shoe_sizes = DataFrame([['Reuven', 46],
                  ['SF', 40],
                  ['AMLF', 40],
                  ['SBLF', 40],
                  ['ADLF', 44]],
                columns=['name', 'shoe_size'])

shoe_sizes
```

```python
(
    ages
    .set_index('name')
    .join(shoe_sizes
        .set_index('name'))
    .groupby('shoe_size')['age'].mean()
)
```

# What about `merge`?

The idea of `merge` is that you can join on any columns, not just on the index.

```python
(
    ages
    .merge(shoe_sizes)
)
```

```python
help(ages.merge)
```

```python
ages.columns
```

```python
shoe_sizes.columns
```

```python
ages.columns.intersection(shoe_sizes.columns)
```

```python
(
    ages
    .merge(shoe_sizes, on='name')   # make the join column explicit
)
```

# What if the names overlap?

Column names in a Pandas data frame must be unique. (Very different from the rows, where uniqueness isn't an issue at all!)

What happens if both of the data frames have one or more columns with the same name? By default, `merge` will add `_x` to the left and `_y` to the right. We can set those, if we want, with keyword arguments.

```python
# Let's make the column names overlap

ages.columns = ['name', 'data']
shoe_sizes.columns = ['name', 'data']
```

```python
ages.merge(shoe_sizes, on='name')
```

```python
(
    ages
    .set_index('name')
    .join(shoe_sizes
        .set_index('name'), lsuffix='_age', rsuffix='_shoe_size')
)
```

```python
help(ages.join)
```

# Strings

So far, we've thought of Pandas and data frames as mostly being useful for working with numbers. And that's true! But they also handle strings (textual data) very well.

NumPy, which is the underlying system for Pandas, is *terrible* when it comes to working with strings. From the beginning, Pandas ignored the NumPy implementation, and did its own thing for strings. That is: The dtype for strings in Pandas was, for a long time, "object". That meant that the value stored in NumPy/Pandas was a pointer (i.e., a memory address) in Python. We used a *Python* string, and referred to its location.

- The good news is that Python strings are super flexible and useful.
- The bad news is that Python strings are in Python memory, and are large. They take a long time to access, and they use lots of memory.

However... as of Pandas 3.0, we don't need to worry about this any more! Strings are now implemented using PyArrow, a new alternative back-end storage system that is *far* faster and slimmer. Now, the dtype looks like `str` (not `object`), and things work much faster.

```python
s = Series('this is a fantastic example of a series'.split())
s
```

```python
# let's say that I want to find the lengths of the words in this series. How can I?

len(s)   # not really... this will give us the length of the series
```

```python
# how can I run the "len" function on every value in our series?
# (remember, using a "for" loop is almost ways a serious mistake.)

# Pandas provides us with a special "accessor" object, ".str", that we can use on a series
# that .str accessor gives us access to lots of string methods... including len!

s.str.len()  # this is the fastest/best way to get the length of each string
```

```python
s.str.contains('e')   # this is similar to 'e' in (each value of) s
```

```python
s.loc[ s.str.contains('e') ]    # this returns the words in s that contain e|
```

The "str" accessor handles several dozen different methods 

- All of Python's string methods
- Some operators that are now methods (e.g., "contains")
- Some other methods from other languages (e.g., R) that were thrown in for good measure
- Some of the methods have the same names as Python methods, but are more powerful -- often supporting regular expressions

```python
# you can even use [] to retrieve a particular character!

s.str[0]  # first letter of each string!
```

```python
s.str[:3]   # first 3 letters of each string
```

```python
text = Series(
           open('alice-in-wonderland.txt')    # get a file object
           .read()                            # get a string with the file's contents
           .split()                           # get a list of strings, broken across whitespace
)

text
```

```python
text.head(20)
```

```python
# what are the 10 most common words in Alice in Wonderland?

text.value_counts().head(10)
```

```python
# what are the 10 most common words in Alice in Wonderland that have > 4 characters?

(
    text
    .loc[ text.str.len() > 4 ]   # keep only words with > 4 characters
    .value_counts()           # count how often each word appears
    .head(10)                 # keep the 10 most common
)
```

```python
s = '   abcd    '
s.strip()  # returns a new string, based on s, but without the leading/trailing whitespace
```

We can pass a string to `str.strip` as an argument. That string contains all of the characters we want to remove from the start and end of the string.

One particularly useful string for removing punctuation is ... `string.punctuation` in the Python standard library.

```python
import string

s = '"Hello!!!!"'
s.strip(string.punctuation)
```

```python
(
    text
    .loc[ text.str.len() > 4 ]   # keep only words with > 4 characters
    .str.strip(string.punctuation)  # apply str.strip, removing leading/trailing punctuation
    .value_counts()           # count how often each word appears
    .head(10)                 # keep the 10 most common
)
```

```python
# let's add a call to str.lower, to force lowercase

(
    text
    .loc[ text.str.len() > 4 ]   # keep only words with > 4 characters
    .str.strip(string.punctuation)  # apply str.strip, removing leading/trailing punctuation
    .str.lower()
    .value_counts()           # count how often each word appears
    .head(10)                 # keep the 10 most common
)
```

```python
text = 'hello out there. I am fine today. Today is a great day.'

s = Series(text.split('.'))
s
```

```python
# what can I do if I want to analyze the words? Now I'm kind of stuck...
# I can use str.split!

(
    s
    .str.split()  # now the values in our series are Python lists!
)
```

```python
# now, how can I work with individual words?
# the solution is the "explode" method, which takes each list element and makes it an element of the *series*


(
    s
    .str.split()  # now the values in our series are Python lists!
    .explode()
    .dropna()
)
```

```python
s = df['description']

(
    s
    .str.split()
    .explode()
    .loc[ lambda s_: s_.str.len() > 4 ]
    .str.lower()
    .str.strip(string.punctuation)
    .loc[ lambda s_: ~s_.isin(['flavors', 'aromas', 'finish', 'drink']) ]
    .value_counts()
    .head(10)
)
```
