# How many data structures?

When we create a series, we're creating a wrapper around a NumPy array. So really, we're creating two data structures.

But actually, a NumPy array is itself two data structures, a Python wrapper and a C-level storage system.

So when we create a series, we really have:

1. The Pandas series
2. The NumPy array's Python layer
3. The NumPy array's C layer

```python
# what if I had used some floats?

s = Series([10, 20.5, 30.7, 40, 50])
s
```

# Retrieving from a series

1. We can use `[]`, but it's a bad idea -- sets up bad habits
2. Better is to use `.loc` with `[]` (not `()`), so that we can retrieve individual elements, multiple elements ("fancy indexing"), or slices.
3. We can also use `.iloc` with `[]` (not `()`), and again can retrieve individual elements, multiple elements with fancy indexing, or slices.

The difference between `.loc` and `.loc` is that the former works on the index, whereas the latter works on the position.

```python
# I can create a series with an index by specifying that at creation time

s = Series([10, 20, 30, 40, 50],
           index=list('abcde'))
s
```

```python
s.loc['b']  # retrieve element b
```

```python
s.iloc[1]  # retrieve element at position 1
```

```python
s.loc[['b', 'd']]  # fancy indexing using .loc and the index
```

```python
s.iloc[[1, 3]]    # fancying indexing using .iloc and the position
```

```python
s.loc['b':'d']   # slice from b to d -- up to AND INCLUDING
```

```python
s.iloc[1:3]   # slice from 1 to 3 -- up to AND NOT INCLUDING (normal in Python)
```

# Methods we can use

Some methods on a series:

- `.min`
- `.max`
- `.mean`
- `.median`
- `.mode`
- `.std` (standard deviation)
- `.sum`
- `.count`
- `.value_counts` -- returns a series showing how often each element appears in the series, from most common to least common
- `.head` -- returns the first 5 elements, unless you specify a different number
- `.tail` -- returns the final 5 elements, unless you specify a different number

# Vector-based calculations

If you have two series, you can perform math operations across them. That means: If you add two series together:

1. They must be of the same length
2. The addition will be performed on each index -- the items with a, the items with b, etc. If you haven't specified an index, then it'll use the default of 0, 1, 2, etc.
3. You get back a new series reflecting the operation you performed.
