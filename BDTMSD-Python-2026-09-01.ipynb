# Modules and imports

If I want to share code that I've written with other people, *or* if I want to use code in a number of different programs, *or* if I want to use code that someone else has written, then I want to use a *module*.

A module is a file containing Python code (basically variable and function definitions) that I can then use inside of my program (or others can use in theirs). This helps us to fulfill the DRY rule -- don't repeat yourself. This has so many advantages!

If there is a module that is already installed on my system, then I can use it in my program with `import`.

## Two versions of `import`:

- `import mymod` -- this loads a module into Python's memory, and makes all of the names it defined available via attributes. So if it defined a variable `x`, we can get it as `mymod.x`. If it defined a function `hello`, then we can invoke it as `mymod.hello()`.
- `from mymod import x` -- the `from .. import` syntax doesn't change what is loaded into memory. But it does change what name(s) are defined. In this example, `mymod` will not be defined as a variable. But `x` will be! This means that we won't be able to say `mymod.x`, but we will be able to say `x`. The idea is that if `x` is a name you're going to use often, then you probably don't want to be bogged down writing its full name.

# More `import` options

There are a few scenarios that people encounter when importing:

- The name is really long, and you would like to have a shorter alias
- The name conflicts with another name in your system, and you want to avoid that duplication/collision
- There might be a standard/convention for naming the module

For any/all of these, you can use the `import ... as` syntax. The idea is you say `import mymod as m`, or whatever alias you want to give it. Once you do that, the variable is set to be `m`; you cannot use the original name `mymod`.

We do this all of the time, especially in the data-analysis world. When we start to use Pandas, we won't say `import pandas`. Rather, we'll say `import pandas as pd`. 

Another, related `import` possibility is if you want to import a particular name from a module, and give it an alias. For that, you can say `from random import randint as ri`. This imports the entire `random` module, but doesn't assign the `random` variable to it. Rather, the only variable assigned is `ri`, which refers to `random.randint` (not that either name is available right now).
