# What the heck? Data analysis with Python?

Python is a great (friendly, useful) programming language. But it is not fast at executing. It is not slim when it comes to memory usage.

How can it be that Python is the #1 language for data analysis, which depends on fast execution and optimizing memory?

Answer: NumPy. This is a data structure that basically is implemented in C (very small, very fast) but has a very thin layer of Python on top of it. We can use it, and benefit from fast, small programs while remaining in the Python world.

NumPy still exists, and is still great.. but it's also very low level.  You lose a lot of functionality that you might want:

- Reading and writing from many different file formats (and databases, etc.)
- Lots of convenience functions for analysis
- Outputting to many different types of plots and graphs

Pandas does all of this, and it does so by basically being a wrapper around NumPy.

Using Pandas, you get the best of all worlds:

- Friendly, helpful, high-level functionality
- Fast execution and small memory footprint thanks to NumPy underneath

Pandas -- Python data analysis library. Polars is written in the Rust language, and its web site is pola.rs. 

- Pandas is not very Pythonic -- by that, I mean that there is more than one way to do just about everything.
- Pandas is ridiculously huge

# Where is Python in all of this?

1. Pandas is written in Python.
2. We'll be using Python strings, ints, and floats, at least to populate our Pandas data structures
3. When we invoke Pandas methods, our arguments will be Python data structures.
4. We will write zero or near-zero functions.
5. We will use almost no `if` statements or `for` loops.
6. We'll rely on a lot of functionality and assumptions in the Pandas data structures, i.e., series and data frames.

# Basic Pandas

Pandas is a Python package (i.e., bunch of modules) that is installed from PyPI. We can load it as

```python
import pandas as pd
from pandas import Series, DataFrame
```

```python
s = Series([10, 20, 30, 40, 50, 60])   # I pass a Python list of integers to Series to create a new series object
```

```python
type(s)  # what kind of value do I have?
```

```python
s
```

People often ask me why I don't call Python lists "arrays." Technically speaking, an array's elements must all be of the same type. Pandas series, because they're based on NumPy arrays, must have all elements contain the same type. In the Pandas world, we call those "dtypes."

The dtype of our series is `int64`, meaning 64-bit integers.



# Summarize

1. We created a new `Series` object by passing a Python list of integers to `Series` (aka `pd.Series`)
2. Our series needs values that are all of the same type. We passed integers, and those were all turned into `int64`, a NumPy dtype.
3. The series can have any number of values in it, but cannot change size.
4. NumPy is always lurking behind the scenes, and we can see it with `s.values`
5. We can see the index on the left side.
6. We can see the dtype at the bottom.

```python
s.values
```

```python
type(s.values)
```

```python
mylist = [10, 20, 30, 40, 50, 60]  # a Python list
```

```python
mylist + mylist  # if I add a list to itself, what do I get?
```

```python
# what if I add s to itself?

s + s
```

The result of adding a series to itself is not a new, longer series! Rather, Pandas adds each of the elements at the same index, and returns a new series -- of the same size, with the same index, but with each value doubled. This is part of its "vectoral computation" system, which is part of its charm/appeal.

```python
# can I retrieve elements from our series?

s[2]
```

```python
s[5]
```

```python
s[-1]  # will this give me the final value?
```

# Two ways to set/retrieve

- We can use `.loc` to set+retrieve values with the index
- We can use `.iloc` to set+retrieve values with the *position*.

Right now, both of these are the same. We haven't changed the index to be anything other than 0-the length.

It's also important to know that `.loc` and `.iloc` use `[]`, not `()`.

```python
s
```

```python
s.loc[0]  # get the item at index 0
```

```python
s.loc[4]  # get the item at index 4
```

```python
s.iloc[0]  # get the item at position 0
```

```python
s.iloc[4]  # get the item at position 4
```

```python
s.describe()
```

```python
s.describe().loc[2]  # this will give us an error, because there is no index 2!
```

```python
s.describe().iloc[2]  # get from position 2
```

```python
# to get the same thing via .loc, I need to use the index -- the actual string that is there!

s.describe().loc['std']
```

```python
s
```

```python
# can I use slices with .loc and .iloc? YES!

s.iloc[2:6]  # from index 2, up to and *NOT INCLUDING* 6
```

```python
# what about with .loc?

s.loc[2:6]  # from index 2, up to AND INCLUDING 6... why?
```

```python
s.describe()
```

```python
s.describe().loc['std':'max']  # if you're using a slice with .loc, it includes the endpoint
```
