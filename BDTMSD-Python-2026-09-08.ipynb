# What the heck? Data analysis with Python?

Python is a great (friendly, useful) programming language. But it is not fast at executing. It is not slim when it comes to memory usage.

How can it be that Python is the #1 language for data analysis, which depends on fast execution and optimizing memory?

Answer: NumPy. This is a data structure that basically is implemented in C (very small, very fast) but has a very thin layer of Python on top of it. We can use it, and benefit from fast, small programs while remaining in the Python world.

NumPy still exists, and is still great.. but it's also very low level.  You lose a lot of functionality that you might want:

- Reading and writing from many different file formats (and databases, etc.)
- Lots of convenience functions for analysis
- Outputting to many different types of plots and graphs

Pandas does all of this, and it does so by basically being a wrapper around NumPy.

Using Pandas, you get the best of all worlds:

- Friendly, helpful, high-level functionality
- Fast execution and small memory footprint thanks to NumPy underneath

Pandas -- Python data analysis library. Polars is written in the Rust language, and its web site is pola.rs. 

- Pandas is not very Pythonic -- by that, I mean that there is more than one way to do just about everything.
- Pandas is ridiculously huge

# Where is Python in all of this?

1. Pandas is written in Python.
2. We'll be using Python strings, ints, and floats, at least to populate our Pandas data structures
3. When we invoke Pandas methods, our arguments will be Python data structures.
4. We will write zero or near-zero functions.
5. We will use almost no `if` statements or `for` loops.
6. We'll rely on a lot of functionality and assumptions in the Pandas data structures, i.e., series and data frames.
