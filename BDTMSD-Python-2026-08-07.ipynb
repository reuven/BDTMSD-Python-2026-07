````python
# the delimiter can be whatever text you want, including multiple characters

s = 'abcd::ef::ghijk::lm'

s.split('::')
````

However, you can only name one string that is your delimiter. You cannot say that sometimes `:` is the delimiter, and sometimes `.` is the delimiter, and it should handle either of those single characters as a delimter.

Much more common is to have text separated by whitespace -- space, newline (\n), carriage return (\r), tab (\t), and vertical tab (\v). `str.split`, if you don't pass it any argument at all, will take any combination and any number of whitespace characters to be the delimiter.

````python
s = 'this is a bunch of words'
s.split(' ')
````

````python
s.split()
````

````python
s = 'this    is   \n\n\n\n\n   a bunch\r\r\r\r\t\t\t of \t\t\v\v\nwords'
s.split(' ')
````

````python
s.split()
````

How can `str.split` be invoked either with 0 arguments (i.e., any whitespace, of any combination and length) or 1 argument (the delimiter)?

This is thanks to *default argument values*. Meaning: If you don't pass a value, `str.split` has its own default waiting to be used. In this case, the default is the special value `None`.

What about the opposite? What about taking a list of strings, and turning that into one string?

For that, we have a separate/different method, `str.join`.

Many languages with a `split`/`join` pair of methods have one or both of them work on the list. But in the case of Python, both are string methods!

- `str.split` is a string method that returns a list of strings
- `str.join` is a string method that returns a new string

If you have a list of strings, and want to get a single string back, you can invoke `str.join`. The big question is: What will be put between the list elements (i.e., the strings)? In other words, what will be the delimiter between fields in our new string?

Whatever you want. I call this string the "glue," beacuse it goes between list elements (and not on the outside).

````python
mylist = ['abcd', 'ef', 'ghij', 'klm']

'*'.join(mylist)  # I invoke the str.join method on the glue string, and the argument is the list!
````

````python
'______'.join(mylist)
````

````python
'\n'.join(mylist)   # printed representation of our string
````

````python
print('\n'.join(mylist))
````

````python
# 1. Remember that it's a string method!

mylist.join('*')  # this seems more logical!
````

````python
# 2. Remember that it must be a list of strings!

mylist = [10, 20, 30, 40, 50]

'*'.join(mylist)  # error -- we can only join a list of strings!
````

Why isn't it a list method?

Answer: Because `str.join` works with any sequence. You can pass it a string, list or tuple, and it'll work just fine.

````python
' '.join('abcde')  # join them with spaces
````

````python
''.join(['abcd', 'ef', 'ghi'])  # join together with the empty string
````

# Exercise: Pig Latin sentence (nicer output)

Repeat the previous exercise, but make sure that the output is all on one line, rather than spread across multiple lines.

https://practice.lernerpython.com/classroom/ce06963a8d/ex-18

````python
# what's wrong with this?

input_string = 'abcd'
output_string = ''

for one_character in input_string:
    output_string += one_character.upper()

output_string    
````

Strings are immutable. You cannot change them, no matter what you do. 

If you use `+=` on a string, you aren't adding to the string. Rather, you are creating a new string, and assigning it back to the same variable.

This means that in the above code, we created 5 strings -- the empty string, then `'A'`, then `'AB'`, then `'ABC'`, then `'ABCD'`. This means that with each iteration, we're creating a new string *and* we're throwing the old string out.

Technically, there isn't an issue with this. But in reality, Python can only take so much of this.

Using `+=` on a string inside of a loop is asking for trouble at some point.

````python
# a better way: use a list, and then join!


input_string = 'abcd'
output_list = []   # start with an empty list

for one_character in input_string:
    output_list.append(one_character.upper())  # lists are designed to be appeneded to

''.join(output_list)  # after creating my output_list, I just run join on it and get the final result
````
