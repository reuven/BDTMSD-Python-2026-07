```python
s = 'abcd.ef.ghij.kl'

s.split('.')   # str.split always returns a new list of strings, based on the original string, using the argument as a delimiter
```

```python
s
```

```python
# to hold onto the list that we created with str.split, we must assign it to a variable

fields = s.split('.')
fields
```

```python
s
```

```python
# you can also just use the result of str.split in some other expression

for one_field in s.split('.'):
    print(one_field)
```

```python
# the delimiter can be whatever text you want, including multiple characters

s = 'abcd::ef::ghijk::lm'

s.split('::')
```

However, you can only name one string that is your delimiter. You cannot say that sometimes `:` is the delimiter, and sometimes `.` is the delimiter, and it should handle either of those single characters as a delimter.
