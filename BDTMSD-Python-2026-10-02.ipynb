# This is grouping

The idea behind grouping is that:

- We want to run an analysis (typically, an aggregation method)
- On one column
- For each distinct value of a categorical column

The syntax is as follows:

- `df.groupby`
- In the `()`, we'll name the categorical -- this is a column name, and there should be relatively few values in this column
- After that, we'll use `[]` to name the data column on which we'll be calculating
- After that, we invoke an aggregation method

```python
#             categorical       data column      aggregation method
df.groupby('passenger_count')['trip_distance'].mean()
```

The returned series has a standard format:

- The index contains the distinct values from our categorical
- The values of the series are the result of invoking our aggregation method on the data column, filtered so that each categorical value gets a turn

```python
# I could be dumb about this, if I want!

# for every distinct total_amount,
# calculate the mean trip_distance

df.groupby('total_amount')['trip_distance'].mean() 
```

# What aggregation methods do we have?

- `min`
- `mean`
- `std`
- `median`
- `max`
- `count` -- how many non-`NaN` values?
- `sum`
- `nunique` -- how many distinct values were there

# What about more than one of each?

The rule of thumb in Pandas is that wherever you can use a single column (i.e., a name/string), you can pass a list of columns (i.e., a list of strings). 



```python
df = pd.read_csv('taxi.csv')
```

```python
# maybe I want the mean for not just trip_distance, but also total_amount, for each passenger_count.

# outer [] are part of the syntax for groupby
# inner [] are to show that we have a list of columns we want to analyze

(
    df
    .groupby('passenger_count')[['trip_distance', 'total_amount']].mean() 
)
```

```python
# maybe I want not just to group on passenger_count, but also on VendorID, the manufacturer.


results = (
    df
    .groupby(['passenger_count', 'VendorID'])['trip_distance'].mean() 
)

# this returns a series with a MULTI-index!

results
```

```python
results.loc[1]  # this returns the subset of rows where passenger_count is 1
```

# more than one aggregation method



```python
# if we want to invoke more than one aggregation method, we use the "agg" method and pass (as individual parameters)
# The aggregation methods, we want to run.

results = (
    df
    .groupby('passenger_count')['trip_distance'].agg(['mean', 'std'])
)

results
```

```python
# Find, for Season + Year, the mean height and weight.

(
    df
             #year+season are category
                                    # height and weight in []
                                        # mean of both height and weight
    .groupby(['Year', 'Season'])[['Height', 'Weight']].mean()
)
```
