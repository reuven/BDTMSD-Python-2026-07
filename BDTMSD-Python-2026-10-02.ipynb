```python
import pandas as pd
```

```python
df = pd.read_csv('taxi.csv')
```

```python
# What is the mean `trip_distance` for 1-passenger rides?

(
    df
    .loc[ pd.col('passenger_count')  == 1 ,   # row selector
        'trip_distance']                      # column selector
    .mean()
)
```

```python
# What is the mean `trip_distance` for 2-passenger rides?

(
    df
    .loc[ pd.col('passenger_count')  == 2 ,   # row selector
        'trip_distance']                      # column selector
    .mean()
)
```

```python
# What is the mean `trip_distance` for 3-passenger rides?

(
    df
    .loc[ pd.col('passenger_count')  == 3 ,   # row selector
        'trip_distance']                      # column selector
    .mean()
)
```

# This is grouping

The idea behind grouping is that:

- We want to run an analysis (typically, an aggregation method)
- On one column
- For each distinct value of a categorical column

The syntax is as follows:

- `df.groupby`
- In the `()`, we'll name the categorical -- this is a column name, and there should be relatively few values in this column
