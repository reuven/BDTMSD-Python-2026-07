# This is grouping

The idea behind grouping is that:

- We want to run an analysis (typically, an aggregation method)
- On one column
- For each distinct value of a categorical column

The syntax is as follows:

- `df.groupby`
- In the `()`, we'll name the categorical -- this is a column name, and there should be relatively few values in this column
- After that, we'll use `[]` to name the data column on which we'll be calculating
- After that, we invoke an aggregation method

```python
#             categorical       data column      aggregation method
df.groupby('passenger_count')['trip_distance'].mean()
```

The returned series has a standard format:

- The index contains the distinct values from our categorical
- The values of the series are the result of invoking our aggregation method on the data column, filtered so that each categorical value gets a turn

```python
# I could be dumb about this, if I want!

# for every distinct total_amount,
# calculate the mean trip_distance

df.groupby('total_amount')['trip_distance'].mean() 
```

# What aggregation methods do we have?

- `min`
- `mean`
- `std`
- `median`
- `max`
- `count` -- how many non-`NaN` values?
- `sum`
- `nunique` -- how many distinct values were there

# What about more than one of each?

The rule of thumb in Pandas is that wherever you can use a single column (i.e., a name/string), you can pass a list of columns (i.e., a list of strings). 



```python
df = pd.read_csv('taxi.csv')
```

```python
# maybe I want the mean for not just trip_distance, but also total_amount, for each passenger_count.

# outer [] are part of the syntax for groupby
# inner [] are to show that we have a list of columns we want to analyze

(
    df
    .groupby('passenger_count')[['trip_distance', 'total_amount']].mean() 
)
```

```python
# maybe I want not just to group on passenger_count, but also on VendorID, the manufacturer.


results = (
    df
    .groupby(['passenger_count', 'VendorID'])['trip_distance'].mean() 
)

# this returns a series with a MULTI-index!

results
```

```python
results.loc[1]  # this returns the subset of rows where passenger_count is 1
```

# more than one aggregation method



```python
# if we want to invoke more than one aggregation method, we use the "agg" method and pass (as individual parameters)
# The aggregation methods, we want to run.

results = (
    df
    .groupby('passenger_count')['trip_distance'].agg(['mean', 'std'])
)

results
```

```python
(
    df
    .groupby(['Year', 'Season'])['Height'].mean()
)
```

This works, and it has the right calculations. But it's very long. Many people prefer to have it in a more rectangular shape.

What does that mean?

Because the index is a *multi-index*, with more than one level, I can send one particular level to be back as a regular column. Or I can send one particular level to be the column names.



```python
(
    df
    .groupby(['Year', 'Season'])['Height'].mean()
    .unstack('Season')  # unstack means: move it from being part of the index to being part of the column
)
```

# Pivot table

A pivot table (in Pandas) is a data frame that you created with the `pivot_table` method. It's basically a 2D groupby:

- One categorical's unique values are used for the index (rows)
- Another categorical's unique values are used for the column names.
- We need a data column
- We need an aggregation method

```python
(
    df
    .pivot_table(index='Year',
                 columns='Season',
                 values='Height',
                 aggfunc='mean')
)
```

```python
(
    df
    .pivot_table(index='Year',
                 columns='Season',
                 values='Height',
                 aggfunc='mean')
    .stack()
)
```

# Where would I use a pivot table?

Anywhere that you have two categorical columns in a data frame.

- Department vs. geography
- Product vs. sales

# Pandas offers

- `pivot_table`
- `pivot` -- this is basically just a reshaping of data, rather than actual calculations

```python
df = pd.read_csv('/Users/reuven/data/olympic_athlete_events.csv')

df
```

```python
# Create a data frame in which the rows are labeled Sport and the columns are Sex. Show the mean height for each intersection.

(
    df
    .groupby(['Sport', 'Sex'])['Height'].mean()
)
```

```python
# index (rows) - Sport
# columns - Sex
# values - Height
# aggfunc - mean

(
    df
    .pivot_table(index='Sport',
                 columns='Sex',
                 values='Height',
                 aggfunc='mean')
)
```

```python
# Do the same thing for Height and Weight. What happens if you pass a list of columns, rather than just one, to pivot_table?


(
    df
    .pivot_table(index='Sport',
                 columns='Sex',
                 values=['Height', 'Weight'],
                 aggfunc='mean')
)
```

```python

(
    df
    .pivot_table(index='Sport',
                 columns='Sex',
                 values=['Height', 'Weight'],
                 aggfunc='mean',
                margins=True)
)
```

```python
# medal counts

(
    df
    .loc[ pd.col('Team').isin(['United States', 'Great Britain', 'Canada', 'France',
                                'Germany', 'Italy', 'Japan'])]
    .pivot_table(index='Team',
                 columns='Medal',
                 values='Season',
                 aggfunc='count',
                margins=True)
)
```
