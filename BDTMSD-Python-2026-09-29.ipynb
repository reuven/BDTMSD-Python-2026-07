# `read_csv` options

If you're going to read a CSV file into Pandas, you can give it many different hints as to how to read the data, and what to do with it. 

Here are a few common options that you might want to use:

- `sep` -- what is the separator between fields? Normally, by default, it's a comma (`','`). It's pretty common for it to also be a tab character, which we write as `'\t'`. It can also be other things, e.g., `'::'`. If you don't get the separator right, you'll almost certainly get a one-column data frame back from `read_csv`, and then it becomes super obvious.
- `usecols` -- a list of columns you want to include from the file. You can either name the columns using the names on the first line (assuming that they are column names), or you can number them, starting at 0, from the first one.
- `header` -- on what line of the file do we get the names of the columns? By default, this is assumed to be 0 (i.e., the first line). But if it's on line 10, then you'll say that, and lines 0-9 will be ignored by `read_csv`. If there is no header in the CSV, you can say `header=None`, and then the columns will only be numbered.
- `index_col` -- what column should be used as the data frame's index? This is basically the same as invoking `set_index` on a particular column after you have created the data frame.
- `names` -- you can pass a list of names, the same length as the number of columns, and these will be used instead of numbers or whatever names were passed. If you use the `names` keyword argument, then `usecols` cannot use names; you need to use numbers.

These are options for `read_csv`, but there are methods to read from a very large number of file formats, and many of them support (some of these) options, as well. 

```python
filename = 'taxi.csv'

df = pd.read_csv(filename)
df
```

```python
# usecols

filename = 'taxi.csv'

df = pd.read_csv(filename,
                usecols=['passenger_count', 'trip_distance', 'total_amount'])
df
```

```python
# usecols + index_col

filename = 'taxi.csv'

df = pd.read_csv(filename,
                usecols=['tpep_pickup_datetime', 'passenger_count', 'trip_distance', 'total_amount'],
                index_col='tpep_pickup_datetime')
df
```

```python
# rename the columns with "names"

filename = 'taxi.csv'

df = pd.read_csv(filename,
                usecols=[1, 3, 4, 18],   # grab columns by numeric index
                header=0,                # we have column names
                names=['pickup_at', 'num_passengers', 'distance', 'amount'],  # we ignore the column names
                index_col='pickup_at')
df
```

# Exercise: Titanic

1. Load `titanic.csv` into a data frame.
2. Find the mean age of people who survived.
3. Find the mean age of men who survived.
4. Find the mean age of women who survived, and got on board in Southampton ("S").

Load the minimum data you need to answer these questions.

https://practice.lernerpython.com/classroom/ce06963a8d/ex-60

```python
filename = 'titanic3.csv'

df = (
    pd
    .read_csv(filename,
              usecols=['survived', 'age', 'sex', 'embarked'])
    .dropna()
)

df
```

```python
# Find the mean age of people who survived.

(
    df
    .loc[ 
        pd.col('survived') == 1   # row selector
        ,
        'age'                     # column selector
        ]
    .mean()
)
```

```python
# mean age of non-survivors

(
    df
    .loc[ 
        pd.col('survived') == 0   # row selector
        ,
        'age'                     # column selector
        ]
    .mean()
)
```

```python
# Find the mean age of men who survived.

(
    df
    .loc[ pd.col('survived') == 1 ]   # first .loc -- row selector for survived
    .loc[ 
        pd.col('sex') == 'male'   # second .loc -- row selector keeping males
        ,
        'age'                     # column selector -- keep the age column
        ]
    .mean()
)
```

```python
# Find the mean age of women who survived, and got on board in Southampton ("S").


(
    df
    .loc[ pd.col('survived') == 1 ]   # first .loc -- row selector for survived
    .loc[ pd.col('embarked') == 'S' ] # second .loc -- only those got on in Southampton
    .loc[ 
        pd.col('sex') == 'female'   # second .loc -- row selector keeping females
        ,
        'age'                     # column selector -- keep the age column
        ]
    .mean()
)
```

# Excel files

Excel files are a bit better than CSV, in that they typically tag their data with appropriate data types, so Pandas doesn't have to work as hard. Also, it can have multiple sheets in a file -- so there will be times when you need to specify what sheet you want. You can use many of the same options as you did with `read_csv`, including -- `header`, `usecols`, and `index_col`. 

Pandas relies on other Python packages to read and load Excel files. If you don't have any installed, then invoking `read_excel` will give an error message, suggesting which package to install.

A big, important option is `sheet_name` -- if there is more than one sheet, then invoking `read_excel` returns the first one, which is often an explanatory sheet. You can name the sheet with a string (its name) or by passing an integer indicating which sheet you want. You can even ask for a list of indexes/names.

```python
df['price'].tail(1) > df['price'].median()
```

# Other file formats

CSV and Excel are definitely the most common/popular file formats. However, Pandas works with a *lot* of other file formats, too:

- JSON -- if you have data in JSON format, you can read it directly into a Pandas data frame.
- Parquet -- this comes from the Apache Arrow open-source project, which is slowly but surely replacing the NumPy back end in Pandas. Parquet format stores the data in binary form, with the dtypes attached. When you load a Parquet file, there's no guesswork as to the dtypes. This is thus more accurate and also far faster. Also, because Parquet is a binary file format, it's also far smaller than a corresponding CSV.

# Sorting our data

It's very common for us to want to sort our data -- less for Pandas, and more for us! We want to eyeball the larget `n` values in a data frame. Or the smallest `n` values, instead.

There are actually *two* basic ways to sort in Pandas, along with a few other, more minor players.

The big methods are:

- `sort_index`
- `sort_values`

Neither of these methods actually changes a data frame. In both cases, invoking the method returns a *new* data frame, identical to the previous one, except that the order of the rows has changed.

```python
df
```

```python
df.sort_index()  # this returns a new data frame, sorted by date (in ascending order)
```

```python
df.sort_index(ascending=False)   # now it'll return a new data frame, sorted by date (in descending order)
```

```python
# it's common to use sort_index along with head/tail

df.sort_index().head(10)  # this will give the 10 rows with the lowest-value indexes
```

```python
# sort_values is a bit more complex -- in a series, we can just invoke sort_values. But in a data frame, we have
# multiple columns. Which do we want to sort by? Even in this case, where we have *one* column, it's still a data frame,
# and we have to indicate it.

df.sort_values('price').dropna()
```
