# Five-minute intro to Jupyter

When we're using Jupyter, we're writing into *cells*. A cell can be in one of two modes:

- *Edit mode*, which means that when I type, what I type goes into Jupyter. To enter edit mode, just click inside of a cell or press Enter.
- *Command mode*, which means that each letter that you type is given to Jupyter as a command to do something, not to type something. To enter command mode, click to the left of a cell or press ESC.

What commands can you give in command mode?

- `x` -- cut the current cell
- `c` -- copy the current cell
- `v` -- paste the most recent cut/copy
- `a` -- create a new cell *above* the current one
- `b` -- create a new cell *below* the current one
- `m` -- make the cell into "Markdown" mode (like right now) for documentation
- `y` -- make the cell into "Python" mode for writing/running code

No matter what, when you're done doing something in Jupyter, and want to run/produce it, just use shift+Enter.

```python
# this is a comment. Comments start with # and go to the end of the line. They can start anywhere.
# Python ignores comments completely. They are for yourself and your colleagues.

print('Hello to everyone out there!')    # this is print, a function -- aka a verb in Python
```

# `print`

`print` is a function. To execute/call/run/invoke the function, you use `()` after its name. In the parentheses, you can put any Python value. For now, we're just using text, which is known as a "string" in Python. Strings can be created with either `''` or `""`. 

Whatever you put inside of `print`'s parentheses is displayed on the screen.

```python
# what about numbers?

print(10)   # here, 10 is an integer, a numeric value, not a text string -- so no quotes around it
```

```python
print('10')  # does this look different?
```

```python
# integers and text strings are different data types

print(2 + 5)  # print only sees the result of 2+5, and displays its result
```

```python
print('2' + '5')  # what if I have two text strings that happen to contain numbers?
```

```python
print('Hello, ' + 'world')   # now we'll print a single greeting to the user
```
