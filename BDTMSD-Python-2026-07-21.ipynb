# Five-minute intro to Jupyter

When we're using Jupyter, we're writing into *cells*. A cell can be in one of two modes:

- *Edit mode*, which means that when I type, what I type goes into Jupyter. To enter edit mode, just click inside of a cell or press Enter.
- *Command mode*, which means that each letter that you type is given to Jupyter as a command to do something, not to type something. To enter command mode, click to the left of a cell or press ESC.

What commands can you give in command mode?

- `x` -- cut the current cell
- `c` -- copy the current cell
- `v` -- paste the most recent cut/copy
- `a` -- create a new cell *above* the current one
- `b` -- create a new cell *below* the current one
- `m` -- make the cell into "Markdown" mode (like right now) for documentation
- `y` -- make the cell into "Python" mode for writing/running code

No matter what, when you're done doing something in Jupyter, and want to run/produce it, just use shift+Enter.

```python
# this is a comment. Comments start with # and go to the end of the line. They can start anywhere.
# Python ignores comments completely. They are for yourself and your colleagues.

print('Hello to everyone out there!')    # this is print, a function -- aka a verb in Python
```

# `print`

`print` is a function. To execute/call/run/invoke the function, you use `()` after its name. In the parentheses, you can put any Python value. For now, we're just using text, which is known as a "string" in Python. Strings can be created with either `''` or `""`. 

Whatever you put inside of `print`'s parentheses is displayed on the screen.

```python
# what about numbers?

print(10)   # here, 10 is an integer, a numeric value, not a text string -- so no quotes around it
```

```python
print('10')  # does this look different?
```

```python
# integers and text strings are different data types

print(2 + 5)  # print only sees the result of 2+5, and displays its result
```

```python
print('2' + '5')  # what if I have two text strings that happen to contain numbers?
```

```python
print('Hello, ' + 'world')   # now we'll print a single greeting to the user
```

```python
# what if I want to combine two values of different types?

print(10 + '20')   
```

# Assignment and variables

So far, we've used numbers, text, and `+` to create and manipulate (and print) values. But this gets boring very quickly. Also, we might want to refer to the same values more than once.

For that, we have *variables*. Each variable is a name that refers to a value. 

In order to give a variable a value, we need to use *assignment*. The assignment operator, as it's known, is `=`. 

**THIS IS NOT THE SAME AS `=` IN MATH!**

If you want to assign a value to a variable, just put the variable on the left side and the value on the right side.

```python
x = 10    # I have now assigned the integer value 10 to the variable x

print(x)
```

# Things to notice and think about

What are the rules for variable names?

- Any combination (almost) of letters, digits, and `_` is legit
    - However, you cannot start with a digit
    - And you should not start or end with `_`
- Capital and lowercase letters are completely different, so `x` and `X` are not the same variables
    - It's traditional in Python to just use lowercase letters
    - Examples: `first_name`
- Use long variable names! The variable names are for you and your colleagues, not for Python

Don't we have to initialize or declare our `x` variable somewhere? Don't we need to tell Python that it'll contain an integer?

No, we don't need to -- the first time you assign to a variable in Python, the variable is created. After that, the variable's value is updated to refer to the new value.

Any variable can refer to any value, of any type.

```python
x = 10
print(x)
```

```python
x = 'abcd'
print(x)
```

In languages where you need to declare variables, it's not just for safety and robustness. It's also because the language needs to know how much memory to allocate for the value, since a variable in those languages is an alias for a location in memory. When the language allocates the memory for an integer vs. a text string, it needs to know what it'll be using, because the memory is different.

In Python, variables are *not* locations in memory. They are what we call "references" to existing values.

More than one variable can refer to the same value. That's not an issue!
