# Five-minute intro to Jupyter

When we're using Jupyter, we're writing into *cells*. A cell can be in one of two modes:

- *Edit mode*, which means that when I type, what I type goes into Jupyter. To enter edit mode, just click inside of a cell or press Enter.
- *Command mode*, which means that each letter that you type is given to Jupyter as a command to do something, not to type something. To enter command mode, click to the left of a cell or press ESC.

What commands can you give in command mode?

- `x` -- cut the current cell
- `c` -- copy the current cell
- `v` -- paste the most recent cut/copy
- `a` -- create a new cell *above* the current one
- `b` -- create a new cell *below* the current one
- `m` -- make the cell into "Markdown" mode (like right now) for documentation
- `y` -- make the cell into "Python" mode for writing/running code

No matter what, when you're done doing something in Jupyter, and want to run/produce it, just use shift+Enter.
