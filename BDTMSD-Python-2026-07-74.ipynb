# Exercise review: Guessing game

1. Define a variable, `secret_number`, which contains an integer.  (Choose a good one!)
2. Print `secret_number` for the user, so that we can debug more easily.
3. Ask the user to guess the number
4. Print one response:
    - Too high
    - Too low
    - You got it

They only have one chance to guess.

```python
secret_number = 72
print(secret_number)

guess = input('Guess: ')   # input always returns a string

# we need to convert guess to be an integer
# easiest way (I think)

guess = int(guess)  # get a new int based on guess, then assign back to guess

if guess == secret_number:
    print('You got it!')
elif guess < secret_number:
    print('Too low!')
else:  # if neither of the first two was True...
    print('Too high!')
```

```python
# this also works (sort of):

guess = int(input('Guess: '))   

# 1. Get input from the user
# 2. Take that input, and feed it to int, getting an integer
# 3. Return that integer to guess

# good news: usually, guess will be an integer
# bad news: if the user enters non-digits, it blows up
```

On line 9, we see

    guess = int(guess)

Several things are going on here:

1. When we use assignment (`=`), it takes the value on the right and assigns it to the variable on the left. It just so happens that the same variable is being used here, but the right-before-left ordering still happens.
2. What happens on the right? We ask for the value of `guess`, and then invoke `int` on that. This returns a new integer based on `guess`. So if `guess` is `'123'`, we'll get the integer `123` back.
3. The value we get (i.e., the new integer) is then assigned to `guess`.

The upshot is that before that line, `guess` is referring to a string. But after that line, `guess` is referring to an integer.

# Integers vs. floats 

Python (and all computers) support at least two different types of numbers:

- `int` (whole number)
- `float` (a number with a decimal point)

Why use one or the other? (Other than the use of a decimal point.)

- Floats are not accurate. Also, they are really slow. And, they are limited in their range.
- Integers might not be able to handle fractional values, but they are super fast and they aren't limited in their size/range.

```python
0.1 + 0.2
```

```python
round(0.2 + 0.2, 2)  # give me the answer to this, but rounded to up to 2 decimal places
```

```python
s = '123'
```

```python
int(s)   # get an integer back
```

```python
float(s)  # get a float back
```

```python
10 == 10.0
```

```python
0.1 + 0.2 == 0.3
```

```python
int(123.456)   # I get an int, which means chopping off the float (not rounding!)
```

# Python's `round` uses "banker's rounding"

We round to the nearest even value. That's to avoid bias if all of our numbers were to round up.

# Strings

We've already talked a bit about strings, which are the type we use in Python to deal with text. There is *no* character type in Python -- a string describes any text, from zero length (known as the "empty string") to one character to as big as you want.

In Python, the system calls a string `str`.

You've already seen that you can define a string with `''` or `""`, and unlike other languages, Python doesn't care which you use.

You can put any characters you want inside of a string; it handles Unicode (i.e., any character from any language or any emoji) can be in the string.

```python
s = 'abcdefghijklmnopqrstuvwxyz'

len(s)  # the "len" function takes a string and returns the number of characters
```

```python
t = 'abc def'
len(t)
```

# What if I have unusual characters in my string?

For example, I might want to have quote characters in my string.

For that, we have a few options:

1. Put a `\` before the quote, and then it is taken literally, rather than used as a delimiter (i.e., marking the start/end of a string)
2. Use the other kind of quote.

```python
s = 'He's very nice'

```

```python
s = "He's very nice"
print(s)
```

```python
# using a backslash here is known as "escaping" 
s = 'He\'s very nice'
print(s)
```

```python
# if I just put "s" into Jupyter, it'll show me the printed representation of the variable
# this is a great programming/programmer technique
# it's not true in non-Jupyter/non-debugger contexts

s
```

```python
s = "She says, "Hello""
print(s)
```

```python
# use '' on the outside

s = 'She says, "Hello"'
print(s)
```

```python
s
```

```python
# what if I have both types of quotes?
# then I need to escape one of them.

s = 'She says, "He\'s very nice"'
print(s)
```
